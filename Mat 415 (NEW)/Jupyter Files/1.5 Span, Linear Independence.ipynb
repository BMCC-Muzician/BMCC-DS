<b>Span</b>
</div></font> 

In this chapter, we continue our discussion of linear combinations. In the previous chapter, we answered the question of whether and how a vector ${\bf y}$ can be expressed as a linear combination of vectors ${\bf v}_1, {\bf v}_2, \cdots, {\bf v}_k$. We now turn to a broader and closely related question: what are **all** the vectors that can be expressed as a linear combination of vectors ${\bf v}_1, {\bf v}_2, \cdots, {\bf v}_k$?


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong> <br> If ${\bf v}_1, {\bf v}_2, \cdots, {\bf v}_k$ are vectors in $\mathbb{R}^n$, then the set of all linear combinations of these vectors is called their <strong>span</strong> and denoted by $span({\bf v}_1, {\bf v}_2, \cdots, {\bf v}_k)$.
</div>


In other words, the span consists of every vector that can be written in the form $$x_1{\bf v}_1 +x_2 {\bf v}_2 +\cdots +x_k {\bf v}_k,$$ where $x_1, x_2, \cdots, x_k$ are real numbers. 

Let's connect the above definition to everything we have discussed so far.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem  </strong> <br> <span style="line-height: 1.6; display: inline-block;"> Let $A =\begin{bmatrix} {\bf a}_1 & {\bf a}_2 &  \cdots & {\bf a}_k \end{bmatrix}$ be an $n \times k$ matrix and let $\bf{b}$ be a vector in $\mathbb{R}^n$. The following statements are equivalent; that is, for a given matrix $A$ and vector ${\bf b}$, they are either all true or all false:</span><br>
    <strong>1. </strong> Vector $\bf{b}$ can be expressed as a linear combinations of ${\bf a}_1, {\bf a}_2,  \cdots, {\bf a}_k$. <br>
    <strong>2. </strong> Vector $\bf{b}$ is a linear combinations of the columns of $A$. <br>
    <strong>3. </strong> Vector  $\bf{b}$ is in the $span({\bf a}_1, {\bf a}_2, \cdots, {\bf a}_k)$.<br>
    <strong>4. </strong> Vector equation $x_1{\bf a}_1+x_2 {\bf a}_2+  \cdots+x_k {\bf a}_k={\bf b}$ is consistent.<br>
    <strong>5. </strong> Matrix equation $A{\bf x}={\bf b}$ is consistent.
</div>


---
**<font color="red">Example 1</font>**

Let $I_3=\begin{bmatrix} 1 & 0 & 0\\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix}$ be the $3 \times 3$ identity matrix and let ${\bf e}_1$, ${\bf e}_2$ and ${\bf e}_3$ be the columns of $I_3$. Show that $span({\bf e}_1, {\bf e}_2,{\bf e}_3) = \mathbb{R}^3$.

**Solution.**

Let ${\bf v}=\begin{bmatrix} v_1 \\ v_2 \\v_3 \end{bmatrix}$ be a vector in $\mathbb{R}^3$. Then $I_3{\bf v}={\bf v}$. So, the equation $I_3{\bf x}={\bf v}$ is consistent and therefore ${\bf v}$ is in the $span({\bf e}_1, {\bf e}_2,{\bf e}_3)$. Since ${\bf v}$ was an arbitrary vector in $\mathbb{R}^3$, every vector in $\mathbb{R}^3$ belongs to the $span({\bf e}_1, {\bf e}_2,{\bf e}_3)$.

---
**<font color="red">Example 2</font>**

Is vector ${\bf b}=\begin{bmatrix}  1 \\ 1 \\ 2\end{bmatrix}$ in the span of vectors ${\bf v}_1=\begin{bmatrix} 1 \\ -2 \\ 1\end{bmatrix}$ and ${\bf v}_2=\begin{bmatrix} 2 \\1 \\ -1\end{bmatrix}$?

**Solution.**

If ${\bf b}$ is in the $span({\bf v}_1, {\bf v}_2)$, then the vector equation $x_1{\bf v}_1+x_2{\bf v}_2={\bf b}$ is consistent. Let's arrange them into an augmented matrix and apply row operations:
$$
\left[
\begin{array}\fbox{\begin{array}{cc|c}
1 & 2 & 1 \\ -2 & 1 & 1 \\ 1 & -1 & 2 \end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_2 +(2) R_1}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cc|c}
1 & 2 & 1 \\ 0 & 5 & 3 \\ 1 & -1 & 2\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_3 +(-1) R_1}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cc|c}
1 & 2 & 1 \\ 0 & 5 & 3 \\ 0 & -3 & 1\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_3 +(\frac{3}{5}) R_2}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cc|c}
1 & 2 & 1 \\ 0 & 5 & 3 \\ 0 & 0 & 14/5\end{array}}\end{array}
\right]$$

Since we have a pivot in the last column of an augmented matrix, this equation is not consistent. Therefore, vector ${\bf b}$ is not in the $span({\bf v}_1, {\bf v}_2)$.

---
From the previous theorem, we can conclude the equation $A{\bf x}={\bf b}$ is consistent for every vector ${\bf b}$ in $\mathbb{R}^m$ if and only if columns of $A$ span $\mathbb{R}^m$. For an equation to be always consistent, however, the augmented matrix $\begin{bmatrix} A | {\bf b} \end{bmatrix}$ cannot have a pivot in the last column no matter what ${\bf b}$ we choose. The only way this could happen is if $A$ already has a pivot in every row.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem  </strong> <br> Let $A$ be an $m \times n$ matrix. Then the following statements are equivalent:<br>
    <strong>1. </strong> Equation $A{\bf x}={\bf b}$ is consistent for every ${\bf b}$ in $\mathbb{R}^m$ <br>
    <strong>2. </strong> Matrix $A$ has a pivot in every row <br>
    <strong>3. </strong> Columns of $A$ span $\mathbb{R}^m$
</div>

---
Let’s look at what the span of vectors looks like from a geometric point of view. We start with the simplest case.

**1.** The span of the zero vector, $span({\bf 0})$. Any scalar multiple of the zero vector is still the zero vector, and the sum of zero vectors is also the zero vector. Therefore, the $span({\bf 0})=\{{\bf 0}\}$, which consists of a single point at the origin.

**2.** Let ${\bf v}$ be a non-zero vector. Taking all scalar multiples of ${\bf v}$ produces a line going through the origin in the direction of ${\bf v}$. The sum of two multiples of ${\bf v}$ is still a multiple of ${\bf v}$:  $c{\bf v}+d{\bf v}=(c+d){\bf v}$. Therefore, $span({\bf v})$ is a line passing through ${\bf v}$ and ${\bf 0}$.

**3.** Let ${\bf v}$ and ${\bf u}$ be non-zero vectors that are not scalar multiples of each other. Then $span({\bf v},{\bf u})$ contains the line through ${\bf v}$ and the origin, the line through ${\bf u}$ and the origin, and all linear combinations of the two vectors. Geometrically, this fills in everything between those lines, forming a plane through the origin.

**4.** Let's take two vectors ${\bf v}$ and ${\bf u}$ from part 3 and add a third vector ${\bf w}$ to this set. If ${\bf w}$ lies in the plane spanned by ${\bf v}$ and ${\bf u}$, then it does not add anything new: $span({\bf v},{\bf u},{\bf w})$ is still the same plane. However, if ${\bf w}$ does not lie in that plane, then the span expands to fill a three-dimensional space, and in this case, $span({\bf v},{\bf u},{\bf w})$ is is a three-dimensional space.

<b>Linear Independence</b>
</div></font> 

When describing the geometric shapes of spans, we repeatedly used phrases such as "nonzero", "not scalar multiples of one another", and "not in the same plane". All of these ideas are captured by a single, fundamental concept: **linear independence**.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition  </strong> <br> A set of vectors $S=\{{\bf v}_1 , {\bf v}_2, \cdots, {\bf v}_k\}$ is <strong>linearly independent</strong> if the vector equation $$x_1{\bf v}_1 +x_2 {\bf v}_2 +\cdots +x_k {\bf v}_k ={\bf 0}$$ has only the trivial solution.<br>

A set that is not linearly independent is <strong>linearly dependent</strong>.
</div>


If the set $S=\{{\bf v}_1 , {\bf v}_2, \cdots, {\bf v}_k\}$ is linearly dependent, then there exist scalars $c_1, c_2, \cdots, c_k$, not all zero, such that $c_1{\bf v}_1 +c_2 {\bf v}_2 +\cdots +c_k {\bf v}_k ={\bf 0}$. Suppose $c_i \neq 0$ for some index $i$. Then we can solve the equation for ${\bf v}_i$:

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}c_1{\bf v}_1 +c_2 {\bf v}_2 +\cdots +c_{i-1}{\bf v}_{i-1}+\hly{c_i{\bf v}_i}+c_{i+1}{\bf v}_{i+1}+\cdots +c_k {\bf v}_k ={\bf 0}$$
$$\hly{c_i{\bf v}_i}=-c_1{\bf v}_1 -c_2 {\bf v}_2 -\cdots -c_{i-1}{\bf v}_{i-1}-c_{i+1}{\bf v}_{i+1}-\cdots -c_k {\bf v}_k$$
$${\bf v}_i=-\frac{c_1}{c_i}{\bf v}_1 -\frac{c_2}{c_i} {\bf v}_2 -\cdots -\frac{c_{i-1}}{c_i}{\bf v}_{i-1}-\frac{c_{i+1}}{c_i}{\bf v}_{i+1}-\cdots -\frac{c_k}{c_i} {\bf v}_k$$

This shows that one of the vectors can be written as a linear combination of the others. This observation leads to another, often more intuitive, way to define linear independence.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition (equivalent form)  </strong> <br> A set $S=\{{\bf v}_1 , {\bf v}_2, \cdots, {\bf v}_k\}$  with at least two vectors is <strong>linearly independent</strong> if no vector in $S$ can be expressed as a linear combination of the remaining vectors.
</div>



---
**<font color="red">Examples of linear dependence</font>**

**1.** The set $S=\{ {\bf 0} \}$ is linearly dependent, since equation $x{\bf 0}={\bf 0}$ can be satisfied by any non-zero $x$.  
**2.** The set $S=\{ {\bf 0}, {\bf v} \}$ is linearly dependent since equation  $x_1{\bf 0}+x_2{\bf v}={\bf 0}$ has many non-zero solutions. For example, $(1){\bf 0}+(0){\bf v}={\bf 0}$.  
**3.** Let  $S=\{ {\bf v}, {\bf u} \}$. Suppose ${\bf v}$ is a multiple of ${\bf u}$ (in other words, they are collinear). For instance,  ${\bf v}=k{\bf u}$. Then $(1){\bf v}-(k){\bf u}={\bf 0}$. Therefore, $S$ is linearly dependent in this case. If neither vector is a scalar multiple of the other, then $S$ is linearly independent.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #4c72b0;
    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
<strong>📌 Note:</strong> This is a very quick way to determine linear dependence of two vectors. If one is a multiple of the other, then vectors are linearly dependent. 
</div>

**4.** If $S=\{ {\bf v}, {\bf u}, {\bf w} \}$ is a set of three vectors lying in the same plane containing the origin (coplanar), then $S$ is linearly dependent.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

    border-left: 4px solid #2c3e50;

    margin: 0.4em 0;
">
<strong>⚠️Caution:</strong> It is possible that three or more vectors are not multiples of one another, but are still linearly dependent. 
</div>

---

If a homogeneous equation $$x_1{\bf v}_1 +x_2 {\bf v}_2 +\cdots +x_k {\bf v}_k ={\bf 0}$$ has only trivial solution, then the corresponding linear system doesn't have any free variables, and therefore, every column of the corresponding coefficient matrix $\begin{bmatrix} {\bf v}_1 & {\bf v}_2 &\cdots & {\bf v}_k\end{bmatrix}$ is a pivot column. 


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong> <br> Let $A$ be an $m \times n$ matrix. Then the following statements are equivalent:  <br>
<strong>1. </strong> $A{\bf x}={\bf 0}$ has only trivial solution.  <br>
<strong>2. </strong> Matrix $A$ has a pivot in every column <br>
<strong>3. </strong> Columns of $A$ are linearly independent.
</div>


Compare this theorem with the analogous theorem about spans.

---
**<font color="red">Example 3</font>**

Let ${\bf v}_1=\begin{bmatrix} 1 \\ 1 \\ 1\end{bmatrix}$, ${\bf v}_2=\begin{bmatrix} 2 \\ 0 \\ -4 \end{bmatrix}$ and ${\bf v}_3=\begin{bmatrix} 0 \\ 4 \\ 12 \end{bmatrix}$. Is the set $\{ {\bf v}_1, {\bf v}_2,{\bf v}_3 \}$ linearly independent? If it is linearly dependent, express one of the vectors in terms of other vectors.

**Solution.**

We combine the given vectors into a single matrix $$\begin{bmatrix} 1 & 2 & 0 \\ 1 & 0 & 4 \\ 1  & -4 & 12 \end{bmatrix}$$

Then we use row reduction to find pivots:

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
\begin{bmatrix} 1 & 2 & 0 \\ 1 & 0 & 4 \\ 1  & -4 & 12 \end{bmatrix}
\stackrel{\color{HotPink}{R_2+(-1)R_1}}{\sim}
\begin{bmatrix} 1 & 2 & 0 \\ 0 & -2 & 4 \\ 1  & -4 & 12 \end{bmatrix}
\stackrel{\color{HotPink}{R_3+(-1)R_1}}{\sim}
\begin{bmatrix} 1 & 2 & 0 \\ 0 & -2 & 4 \\ 0  & -6 & 12 \end{bmatrix}
\stackrel{\color{HotPink}{R_3+(-3)R_2}}{\sim}
\begin{bmatrix} \hl{1} & 2 & 0 \\ 0 & \hl{-2} & 4 \\ 0  & 0 & 0 \end{bmatrix}$$

We have a column without a pivot; therefore, the set $\{ {\bf v}_1, {\bf v}_2,{\bf v}_3 \}$ is linearly dependent. Let's express one of the vectors in terms of the others. We continue with row reduction to find a non-trivial solution.

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
\begin{bmatrix} 1 & 2 & 0 \\ 0 & -2 & 4 \\ 0  & 0 & 0 \end{bmatrix}
\stackrel{\color{HotPink}{R_1+(1)R_2}}{\sim}
\begin{bmatrix} 1 & 0 & 4 \\ 0 & -2 & 4 \\ 0  & 0 & 0 \end{bmatrix}
\stackrel{\color{HotPink}{-\frac{1}{2}R_2}}{\sim}
\begin{bmatrix} 1 & 0 & 4 \\ 0 & 1 & -2 \\ 0  & 0 & 0 \end{bmatrix}$$

Let $x_3=t$, then

$$
\left\{\begin{array}{l}
x_1  & =& -4t \\
x_2 & =& 2t\\
x_3  &=&  t
\end{array}\right.
$$

Any choice of $t$ leads to a solution. Since we would like a non-trivial solution, we should pick a non-zero value for $t$. Let $t=1$. Then 
$$
\left\{\begin{array}{l}
x_1  & =& -4 \\
x_2 & =& 2\\
x_3  &=&  1
\end{array}\right.
$$

And so, $$(-4){\bf v}_1+(2){\bf v}_2+(1){\bf v}_3={\bf 0}$$

Solving this for ${\bf v}_3$ produces:

$${\bf v}_3=4{\bf v}_1-2{\bf v}_2$$


---
**<font color="red">Example 4</font>**

Let ${\bf v}_1=\begin{bmatrix} 1 \\ 1 \\ 1\end{bmatrix}$, ${\bf v}_2=\begin{bmatrix} 2 \\ 0 \\ -4 \end{bmatrix}$ and ${\bf v}_3=\begin{bmatrix} 0 \\ 4 \\ 12 \end{bmatrix}$ (the same as in Example 3). What does $span\left( {\bf v}_1, {\bf v}_2,{\bf v}_3 \right)$ look like?

**Solution.**

At the end of Example 3, we have determined that $${\bf v}_3=4{\bf v}_1-2{\bf v}_2$$ This means that vector ${\bf v}_3$ is already in the $span\left( {\bf v}_1, {\bf v}_2 \right)$ and so, it doesn't add anything new. Therefore, $$span\left( {\bf v}_1, {\bf v}_2,{\bf v}_3 \right)=span\left( {\bf v}_1, {\bf v}_2 \right)$$

Vectors ${\bf v}_1$ and ${\bf v}_2$ are linearly independent. There are a couple of ways to determine this. One way is from the fact that they are not multiples of one another. The second way is to examine the echelon form of the matrix in Example 3. The first two columns (the ones corresponding to ${\bf v}_1$ and ${\bf v}_2$) have pivots and if we consider only the submatrix consisting of those two columns, then every column has a pivot.

Two linearly independent vectors determine a plane. Therefore, the $span\left( {\bf v}_1, {\bf v}_2,{\bf v}_3 \right)$ is a plane containing the origin.

---

The previous example provides a useful way to determine the geometric shape of a span. Suppose $A=\begin{bmatrix} {\bf a}_1 & {\bf a}_2 & \cdots &{\bf a}_n \end{bmatrix}$ and let $B=\begin{bmatrix} {\bf b}_1 & {\bf b}_2 & \cdots &{\bf b}_k \end{bmatrix}$ be the matrix consisting of all pivot columns of $A$. Then $$span\left( {\bf a}_1, {\bf a}_2, \cdots, {\bf a}_n \right)=span\left( {\bf b}_1, {\bf b}_2, \cdots,{\bf b}_k \right)$$

Since $B$ consists of pivot columns of $A$, the columns of $B$ form a linearly independent set and so we can determine what it looks like geometrically.

    border-top: 2px solid rgba(144,62,80,0.3);

    border-radius: 8px;
">
<b>Solutions to Homogeneous Equations</b>
</div></font> 

The augmented matrix for a homogeneous equation has the form $\begin{bmatrix} A | {\bf 0}\end{bmatrix}$, where the last column consists entirely of zeros. Elementary row operations cannot change a column of zeros. As a result, after row reduction, the augmented matrix will still have zeros in the last column. This means that the solution contains no constant terms. Consequently, the solution set of a homogeneous system, when written in vector parametric form, is always a linear combination of vectors.

---
**<font color="red">Example 5</font>**

Let's solve $A{\bf x}={\bf 0}$ when $A=\begin{bmatrix} 1 & 3 & 3 & -2 \\ 2 & 7 & 4 & 0 \end{bmatrix}$

**Solution.**
$$
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 3 & 3 & -2 & 0\\ 2 & 7 & 4 & 0 & 0\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_2 +(-2) R_1}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 3 & 3 & -2 & 0\\  0 & 1 & -2 & 4 & 0\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_1 +(-3) R_2}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 0 & 9 & -14 & 0\\ 0 & 1 & -2 & 4 & 0\end{array}}\end{array}
\right]$$

So, we have two free variables, $x_3$ and $x_4$. Let $x_3=t$ and $x_4=s$. Then, the solution is

$${\bf x}= 
\begin{bmatrix}
x_1 \\
x_2\\
x_3\\
x_4
\end{bmatrix}=
\begin{bmatrix}
-9t &+14s \\
2t &-4s\\
 t& \\
 s&
\end{bmatrix}= 
t\begin{bmatrix}
-9  \\
2\\
 1\\
 0
\end{bmatrix}+
s\begin{bmatrix}
14\\
-4\\
0\\
 1
\end{bmatrix}$$

Since $t$ and $s$ can be any real numbers, the solution can also be expressed as 
$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
span\left(\begin{bmatrix}
-9  \\
2\\
 \hl{1}\\
 \hly{0}
\end{bmatrix},
\begin{bmatrix}
14\\
-4\\
\hl{0}\\
 \hly{1}
\end{bmatrix}\right)$$

---

If we focus on the highlighted parts in the answer for the above example, we can see that the two vectors obtained are linearly independent. Each free variable corresponds to exactly one vector in the solution set: in that vector, the entry associated with its free variable is $1$, while the corresponding entries in all other vectors are $0$. Since it is impossible to obtain $1$ as a linear combination of zeros, none of these vectors can be written as a linear combination of the others. Therefore, the vectors are linearly independent. Geometrically, this means that the solution set of the above system is a plane passing through the origin.

Let's summarize our findings:

Let $A$ be a matrix. Then 

**1.** If $A$ has no free variables, then a homogeneous equation $A{\bf x}={\bf 0}$ has only trivial solution.   
**2.** If $A$ has $k>0$ free variables, then the solution set to a homogeneous equation $A{\bf x}={\bf 0}$ is a span of $k$ linearly independent vectors.

    border-radius: 8px;
">
<b>Solutions to Non-homogeneous Equations</b>
</div></font> 

If the equation $A{\bf x}={\bf b}$ is consistent and non-homogeneous, then the last column of the augmented matrix is not a zero vector. As a result, the solution will include a constant vector that does not depend on any free variables. Moreover, because elementary row operations do not interchange columns, the parts of the solution corresponding to the free variables are exactly the same as those obtained from the homogeneous system $A{\bf x}={\bf 0}$. Therefore, the solution to a non-homogeneous system can always be written as a sum of a constant vector (called **particular solution**) and a linear combination of vectors that satisfy the corresponding homogeneous equation.

Let's use the same matrix $A$ as in Example 5 to solve a non-homogeneous equation.

---
**<font color="red">Example 6</font>**

Solve $A{\bf x}={\bf b}$ when $A=\begin{bmatrix} 1 & 3 & 3 & -2 \\ 2 & 7 & 4 & 0 \end{bmatrix}$ and ${\bf b}=\begin{bmatrix} 2 \\2 \end{bmatrix}$

**Solution.**

$$
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 3 & 3 & -2 & 2\\ 2 & 7 & 4 & 0 & 2\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_2 +(-2) R_1}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 3 & 3 & -2 & 2\\  0 & 1 & -2 & 4 & -2\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_1 +(-3) R_2}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{cccc|c}
1 & 0 & 9 & -14 & 8\\ 0 & 1 & -2 & 4 & -2\end{array}}\end{array}
\right]$$

As before, let $x_3=t$ and $x_4=s$. Then the solution is
$${\bf x}= 
\begin{bmatrix}
x_1 \\
x_2\\
x_3\\
x_4
\end{bmatrix}=
\begin{bmatrix}
8 & -9t &+14s \\
-2 & +2t &-4s\\
 &t& \\
&& s
\end{bmatrix}= 
\begin{bmatrix}
8  \\
-2\\
 0\\
 0
\end{bmatrix}+
t\begin{bmatrix}
-9  \\
2\\
 1\\
 0
\end{bmatrix}+
s\begin{bmatrix}
14\\
-4\\
0\\
 1
\end{bmatrix}$$

--- 
As we can see, the solution set to this non-homogeneous equation consists of a particular solution plus the solution set of the corresponding homogeneous equation. All three vectors appearing in the solution are still linearly independent. In fact, this can be verified by examining the same entries we used in the homogeneous case.

<strong>🏛️ Theorem </strong> <br>Suppose equation $A{\bf x}={\bf b}$ is consistent and let ${\bf p}$ be one particular solution. Then the solution set for $A{\bf x}={\bf b}$ consists of all vectors of the form ${\bf x} = {\bf p}+{\bf h}$, where ${\bf h}$ is any solution to the homogeneous equation $A{\bf x}={\bf 0}$.
</div>

<b>Geometric View of the Solution Set</b>
</div></font> 

The solution sets of homogeneous equations are geometric objects that pass through the origin: they may consist of a single point (the origin), a line through the origin, a plane through the origin, and so on. This is because the solution set of a homogeneous equation is a span of certain vectors. In contrast, the solution set of a non-homogeneous equation is obtained by adding a fixed vector to every vector in the solution set of the corresponding homogeneous equation. Geometrically, this operation can be interpreted as a translation, or a parallel shift, of the original object.

For example, suppose the solution has the form ${\bf x}={\bf p}+t{\bf v}$. Then geometrically, it is a line parallel to a line ${\bf x}=t{\bf v}$ and goes through a point ${\bf p}$.

<center> <img src="data:image/png;base64,  
 " width="40%"> </center>

The expression  ${\bf x}={\bf p}+t{\bf v}$ is the vector parametric form of the equation of a line. More specifically, it is the **equation of a line through ${\bf p}$ parallel to ${\bf v}$**.

Similarly, the **equation of a plane through ${\bf p}$ parallel to ${\bf v}$ and ${\bf u}$** is given by ${\bf x}={\bf p}+t{\bf v}+s{\bf u}$.

```python
import plotly.graph_objects as go
import numpy as np

# Define the domain
xp = np.linspace(-0.5, 4, 50)
yp = np.linspace(-0.5, 4, 50)
X, Y = np.meshgrid(xp, yp)

# x-,y- and z-coordinates of points v, u, w and 0, and the colors for them.
x = [1, 2, 3, 0]
y = [2, 1, 3, 0]
z = [3, 1, 4, 0]
color=['red','red','blue','black']

# Define the plane equation
Z = (5*Y-X)/3

fig = go.Figure()

# Draw plane
fig.add_trace(go.Surface(
    x=xp, y=yp, z=Z,
    colorscale=[[0, '#70D27D'], [1, '#70D27D']],
    opacity=0.4,
    showscale=False
))

# Plot the points
fig.add_trace(go.Scatter3d(
    x=x,
    y=y,
    z=z,
    mode='markers',
    marker=dict(size=5, color=color)
))

# Plot the vector segments
fig.add_trace(go.Scatter3d(x=[0,1],y=[0,2],z=[0,3],mode='lines', line=dict(color='red', width=5)))
fig.add_trace(go.Scatter3d(x=[0,2],y=[0,1],z=[0,1],mode='lines', line=dict(color='red', width=5)))
fig.add_trace(go.Scatter3d(x=[0,3],y=[0,3],z=[0,4],mode='lines', line=dict(color='blue', width=5)))

       
# Set the figure size and axis colors
fig.update_layout(
    width=700,
    height=700,
    showlegend=False,
    scene=dict(
        aspectmode='cube',

        xaxis=dict(
            showbackground=True, showgrid=True,title='',
            backgroundcolor='rgba(220,210,195,0.92)',
            showline=True, linecolor='rgba(120,120,120,0.4)', linewidth=1,
            ticks='', zeroline=False,showticklabels=False
        ),
        yaxis=dict(
            showbackground=True, showgrid=True,title='',
            backgroundcolor='rgba(205,195,180,0.92)',
            showline=True, linecolor='rgba(120,120,120,0.4)', linewidth=1,
            ticks='', zeroline=False,showticklabels=False
        ),
        zaxis=dict(
            showbackground=True, showgrid=True,title='',
            backgroundcolor='rgba(185,175,160,0.95)',
            showline=True, linecolor='rgba(120,120,120,0.4)', linewidth=1,
            ticks='', zeroline=False,showticklabels=False
        )
    )
)

# Draw lines along u and v
xl=np.linspace(-0.2, 2, 50)
yl=2*xl
zl=3*xl
fig.add_trace(go.Scatter3d(x=xl, y=yl, z=zl, mode='lines', opacity=0.7,
                           line=dict(color='#D93319', width=4,dash='longdashdot')))
xl=np.linspace(-0.3, 4, 50)
yl=xl/2
zl=xl/2
fig.add_trace(go.Scatter3d(x=xl, y=yl, z=zl, mode='lines', opacity=0.7,
                           line=dict(color='#D93319', width=4,dash='longdashdot')))


#Draw the boundary of the plane
xb = np.array([-0.5,  4,  4, -0.5, -0.5])
yb = np.array([-0.5, -0.5,  4,  4, -0.5])
zb3 = (5*yb-xb)/3

fig.add_trace(go.Scatter3d(
    x=xb, y=yb, z=zb3,
    mode='lines',
    line=dict(color='#1F2A6A', width=2),
    showlegend=False
))

# Adding the text
text=["<b>v</b>", "<b>u</b>", "<b>w</b>", "<b>0</b>"]
fig.add_trace(go.Scatter3d(
    x=x,
    y=y,
    z=z,
    text=text,
    mode='text', 
    textfont=dict(size=20, color=color)
))
fig.show()
```
