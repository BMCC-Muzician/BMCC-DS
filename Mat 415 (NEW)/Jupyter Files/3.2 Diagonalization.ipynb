<b>Similarity</b>
</div></font> 

The main focus of this chapter is on expressions of the form $P^{-1}AP$, where $P$ is an invertible matrix and $A$ is a square matrix. We can view this as a transformation that changes matrix $A$ into $P^{-1}AP$. Such transformations are called **similarity transformations**. They are important because they preserve many properties of the matrix $A$.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> If $A$ and $B$ are square matrices, then $A$ is <strong>similar</strong> to $B$ if there exists an invertible matrix $P$ such that $B=P^{-1}AP$.
</div>

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #4c72b0;
    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
<strong>📌 Note:</strong> If $A$ is similar to $B$, then $B$ is also similar to $A$. Therefore, we can simply say that $A$ and $B$ are similar.
</div>

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #4c72b0;
    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
<strong>📌 Note:</strong> If we let $Q=P^{-1}$, we can also rewrite the similarity relation as $B=QAQ^{-1}$
</div>



Similar matrices share many properties. In a sense, they can be thought of as the same linear transformation viewed from different coordinate systems.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> If $A$ and $B$ are similar, then <br>
<strong>1.</strong> $det(A)=det(B)$<br>
<strong>2.</strong> $A$ is invertible if and only if $B$ is invertible.<br>
<strong>3.</strong> $A$ and $B$ have the same rank.<br>
<strong>4.</strong> $A$ and $B$ have the same characteristic polynomial and therefore, the same eigenvalues.<br>
<strong>5.</strong> For each eigenvalue, the corresponding eigenspaces of $A$ and $B$ have the same dimension.
</div>




Since diagonal matrices are simple, we can consider matrices that are similar to a diagonal matrix. 

---
**<font color="red">Example 1</font>**

Let $D=\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}$ and $P=\begin{bmatrix} 1 & 1 \\ 1 & 2\end{bmatrix}$. Let $A=PDP^{-1}$. Compute $D^2$, $D^3$, $D^n$ and $A^n$.

**Solution**

$$D^2=\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}=\begin{bmatrix} 2^2 & 0\\0& 3^2\end{bmatrix}$$

$$D^3=\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}\begin{bmatrix} 2 & 0\\0& 3\end{bmatrix}=\begin{bmatrix} 2^3 & 0\\0& 3^3\end{bmatrix}$$

$$D^n=\begin{bmatrix} 2^n & 0\\0& 3^n\end{bmatrix}$$

To find $A^n$, we rewrite it in terms of $D$ and $P$. Notice that in this case, all $P$ and $P^{-1}$, except the outer ones, will cancel.

$$A^n=(PDP^{-1})^n=(PD\underbrace{P^{-1})(P}_{I}D\underbrace{P^{-1})(P}_{I}DP^{-1}) \cdots (PDP^{-1})=PDD \cdots DP^{-1}=PD^nP^{-1}$$

And so, 

$$A^n=PD^nP^{-1}=\begin{bmatrix} 1 & 1 \\ 1 & 2\end{bmatrix}\begin{bmatrix} 2^n & 0\\0& 3^n\end{bmatrix}\begin{bmatrix} 2 & -1 \\ -1 & 1\end{bmatrix}=\begin{bmatrix} 2\cdot 2^n-3^n &  -2^n+3^n \\ 2 \cdot 2^n -2\cdot 3^n & -2^n+2\cdot 3^n\end{bmatrix}$$

---

Notice how easy it is to compute powers of a diagonal matrix. We used this to our advantage to find the $n$-th power of a matrix that is similar to a diagonal matrix. If we had first computed $A$ and then tried to calculate $A^n$ directly, the computation would have been much more complicated.

<b>Diagonalization</b>
</div></font> 

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

">
<strong>🏛️ Definition</strong><br> A matrix $A$ is called <strong>diagonalizable</strong> if it is similar to some diagonal matrix $D$. That is, there exists an invertible matrix $P$ such that $D=P^{-1}AP$. In this case, we say that $P$ <strong>diagonalizes</strong> $A$.
</div>

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
<strong>📌 Note:</strong> $D=P^{-1}AP$ can also be rewritten as $A=PDP^{-1}$, which is more useful in computations. 
</div>
 

The next theorem gives us a criterion for determining whether a matrix is diagonalizable and explains how to diagonalize it.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> An $n \times n$ matrix $A$ is diagonalizable if and only if it has $n$ linearly independent eigenvectors. In fact, if $D=P^{-1}AP$, then the columns of $P$ are linearly independent eigenvectors of $A$, and the diagonal entries of $D$ are the corresponding eigenvalues.
</div>




 **Proof.**

To avoid dealing with inverses we rewrite $D=P^{-1}AP$ as $$ PD=AP$$

First, note that if $P=\begin{bmatrix} {\bf v}_1 & {\bf v}_2 &  \cdots & {\bf v}_n\end{bmatrix}$ and $D$ is a diagonal matrix with diagonal entries $\lambda_1, \cdots, \lambda_n$, then

$$AP=A \begin{bmatrix} {\bf v}_1 & {\bf v}_2 &  \cdots & {\bf v}_n\end{bmatrix} = \begin{bmatrix} A{\bf v}_1 & A{\bf v}_2 &  \cdots & A{\bf v}_n\end{bmatrix}$$

and 

$$PD=\begin{bmatrix} {\bf v}_1 & {\bf v}_2 &  \cdots & {\bf v}_n\end{bmatrix} \begin{bmatrix} \lambda_1 & 0  & \cdots & 0 \\ 0 & \lambda_2 & \cdots & 0 \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \cdots & \lambda_n\end{bmatrix}=\begin{bmatrix} \lambda_1{\bf v}_1 & \lambda_2{\bf v}_2 &  \cdots & \lambda_n{\bf v}_n\end{bmatrix}$$

Since this is an "if and only if" statement, we must prove both directions.

<strong>1.</strong> If $A$ diagonalizable, then $A$ has $n$ linearly independent eigenvectors.


If $A$ is diagonalizable, then there exists invertible matrix $P$ and a diagonal matrix $D$ such that $AP=PD$. Using the computations above, we obtain 

$$\begin{bmatrix} A{\bf v}_1 & A{\bf v}_2 &  \cdots & A{\bf v}_n\end{bmatrix}=\begin{bmatrix} \lambda_1{\bf v}_1 & \lambda_2{\bf v}_2 &  \cdots & \lambda_n{\bf v}_n\end{bmatrix}$$

Therefore, for each $i=1,\cdots,n$, $$A{\bf v}_i=\lambda_i{\bf v}_i$$

Thus, each ${\bf v}_i$ is an eigenvector of $A$ with eigenvalue $\lambda_i$.

Since $P$ is invertible, its columns are linearly independent. Therefore, $A$ has $n$ linearly independent eigenvectors.

<strong>2.</strong> If $A$ has $n$ linearly independent eigenvectors, then $A$ is diagonalizable.

Suppose $A$ has $n$ linearly independent eigenvectors. Then we can construct and an invertible matrix $P$ whose columns are these eigenvectors and a diagonal matrix $D$ whose diagonal entries are the corresponding eigenvalues. Since for each $i$, $A{\bf v}_i=\lambda_i{\bf v}_i$, the computations above imply that $AP=PD$. And so, $A$ is diagonalizable. 

---

Above theorem gives us an algorithm for diagonalizing an $n \times n$ matrix $A$:


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Diagonalization Algorithm </strong><br> 
<strong>1.</strong> Find the eigenvalues of $A$.<br>
<strong>2.</strong> For each eigenvalue, find a basis for its eigenspace. Use the basis vectors as eigenvectors. If the total number of eigenvectors is less than $n$, then $A$ is not diagonalizable and we can stop. Otherwise, continue.<br>
<strong>3.</strong> Form the matrix $P$, whose columns are the chosen eigenvectors.<br>
<strong>4.</strong> Form the diagonal matrix $D$, whose diagonal entries are the corresponding eigenvalues, written in the same order as the eigenvectors in $P$.
</div>



---
**<font color="red">Example 2</font>**

Diagonalize $A=\begin{bmatrix} 0 & 0 & -2 \\ 1 & 2 & 1 \\ 1 & 0 & 3\end{bmatrix}$

**Solution**

**Step 1.** Eigenvalues.

Compute the characteristic polynomial:

$$|A-\lambda I|=\begin{vmatrix} -\lambda & 0 & -2 \\ 1 & 2-\lambda & 1 \\ 1 & 0 & 3-\lambda \end{vmatrix} = (2-\lambda) \begin{vmatrix} -\lambda &  -2  \\ 1  & 3-\lambda \end{vmatrix} = (2-\lambda) (-\lambda (3-\lambda)- (-2))=(2-\lambda)(\lambda^2-3\lambda+2)=(2-\lambda)^2(1-\lambda)=0$$

So, we have two distinct eigenvalues $\lambda=1$ with multiplicity $1$ and $\lambda=2$ with multiplicity $2$.

**Step 2.** Eigenspaces

For each eigenvalue we find a basis for the corresponding eigenspace. 

For $\lambda=1$ we have 

$$A-1 I= \begin{bmatrix} -1 & 0 & -2 \\ 1 & 1 & 1 \\ 1 & 0 & 2 \end{bmatrix}\sim \cdots \sim \begin{bmatrix} 1 & 0 & 2 \\ 0 & 1 & -1 \\ 0 & 0 & 0 \end{bmatrix}$$

The general solution for $(A-1 I){\bf x}={\bf 0}$ is $${\bf x} = t\begin{bmatrix} -2 \\ 1 \\ 1 \end{bmatrix}$$
Therefore, a basis for the eigenspace corresponding to $\lambda=1$ is $$\left\{\begin{bmatrix} -2 \\ 1 \\ 1 \end{bmatrix}\right\}$$

For $\lambda=2$ we have  
$$A-2 I= \begin{bmatrix} -2 & 0 & -2 \\ 1 & 0 & 1 \\ 1 & 0 & 1 \end{bmatrix}\sim \cdots \sim \begin{bmatrix} 1 & 0 & 1 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}$$

Here we have two free variables, so general solution for  $(A-2 I){\bf x}={\bf 0}$ is $${\bf x} = t\begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}+s\begin{bmatrix} -1 \\ 0 \\ 1 \end{bmatrix}$$

Hence, a basis for the eigenspace corresponding to $\lambda=2$ is

$$\left\{ \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}, \begin{bmatrix} -1 \\ 0 \\ 1 \end{bmatrix} \right\}$$

We have three linearly independent eigenvectors, therefore matrix $A$ is diagonalizable.

**Step 3.** Matrix $P$

Columns of matrix $P$ are the three eigenvectors that formed the bases for eigenspaces:

$$P=\begin{bmatrix} -2 & 0&-1\\ 1 & 1&0\\1 &0 &1 \end{bmatrix}$$

**Step 4.** Matrix $D$

The diagonal entries of $D$ are the eigenvalues corresponding to the eigenvectors in $P$, written in the same order:

$$D=\begin{bmatrix} 1 & 0 & 0 \\ 0 & 2 & 0 \\ 0& 0& 2\end{bmatrix}$$


Finally, let's check if $P$ indeed diagonalizes $A$.

$$PDP^{-1}=\begin{bmatrix} -2 & 0&-1\\ 1 & 1&0\\1 &0 &1 \end{bmatrix}\begin{bmatrix} 1 & 0 & 0 \\ 0 & 2 & 0 \\ 0& 0& 2\end{bmatrix}\begin{bmatrix} -1 & 0 & -1 \\ 1 & 1 & 1 \\ 1& 0& 2\end{bmatrix}= \begin{bmatrix} 0 & 0 & -2 \\ 1 & 2 & 1 \\ 1 & 0 & 3\end{bmatrix} =A$$

---


**<font color="red">Example 3</font>**

Diagonalize $A=\begin{bmatrix} 1 & 1 \\ 0 & 1\end{bmatrix}$

**Solution**

This is a triangular matrix, so its eigenvalues are easy to determine. In fact, we have only one eigenvalue $\lambda=1$. It has multiplicity $2$.

To find a basis for the eigenspace, we solve $(A-1 I){\bf x}={\bf 0}$:

$$A-1 I=\begin{bmatrix} 0 & 1 \\ 0 & 0\end{bmatrix}$$

The general solution is $${\bf x}=t\begin{bmatrix} 1 \\0\end{bmatrix}$$

Therefore, the eigenspace for $\lambda=1$ has dimension $1$ and there are no other eigenspaces. It is impossible to find two linearly independent eigenvectors. Therefore, $A$ is not diagonalizable.

---



Recall that eigenvectors corresponding to distinct eigenvalues are linearly independent. Therefore, if an $n \times n$ matrix has $n$ distinct eigenvalues, it must be diagonalizable. However, this condition is sufficient but not necessary for diagonalizability. A matrix may still be diagonalizable even if some eigenvalues are repeated (see Example 2).

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> Suppose $\lambda$ is an eigenvalue of $A$ with algebraic multiplicity $m$. Then the dimension of the eigenspace corresponding to $\lambda$ is less than or equal to $m$.
</div>


This theorem implies that if the dimension of an eigenspace is smaller than the multiplicity of its corresponding eigenvalue, then we will not have enough linearly independent eigenvectors to diagonalize the matrix. Therefore, the matrix will not be diagonalizable.


Another possible obstruction comes from complex eigenvalues. A polynomial with real coefficients may have complex zeros, so a matrix with real entries may have complex eigenvalues. Complex eigenvalues can be associated with rotational behavior. However, in this course, we consider diagonalization only over the real numbers, so we require all eigenvalues to be real. Therefore, if the characteristic polynomial cannot be factored completely into linear factors over the real numbers, we consider the matrix to be non-diagonalizable.

These are the only obstructions to diagonalizability.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br><span style="line-height: 1.5; display: inline-block;"> Let $A$ be an $n \times n$ matrix with distinct real eigenvalues $\lambda_1, \lambda_2 ,\cdots, \lambda_p$.
Then, $A$ is diagonalizable if and only if the sum of the dimensions of its eigenspaces equals $n$. <br> 
Equivalently, $A$ is diagonalizable if and only if:</span><br>
<strong>1.</strong> the characteric polynomial factors completely into linear factors, and <br>
<strong>2.</strong> the dimension of each eigenspace equals the algebraic multiplicity of its corresponding eigenvalue.
</div>


Suppose $A$ is an $n \times n$ diagonalizable matrix, and let $\mathcal{B}_i$ denote a basis corresponding to eigenvalue $\lambda_i$. Then, the union of all the bases $\mathcal{B}_i$ forms a basis for $\mathbb{R}^n$. Such a basis is called an **eigenbasis** for $\mathbb{R}^n$ relative to $A$.

<b>Eigenvectors and Transformations </b>
</div></font> 

For this part, we recall two important facts:

1. Given a linear transformation $T:\mathbb{R}^n \rightarrow \mathbb{R}^n$, the standard matrix for $T$ is a matrix $A$ whose columns are images of the standard basis vectors. In this case, $T$ can be expressed as a matrix transformation $T({\bf x})=A{\bf x}$.

2. Given a basis $\mathcal{B}$ for $\mathbb{R}^n$, there exists a change-of-basis matrix $P_{\mathcal{B}}$ such that ${\bf x}=P_{\mathcal{B}}[{\bf x}]_{\mathcal{B}}$, where $[{\bf x}]_{\mathcal{B}}$ is a coordinate vector of $\bf x$ relative to the basis $\mathcal{B}$. The columns of $P_{\mathcal{B}}$ are precisely the vectors in the basis $\mathcal{B}$.

---

Suppose $A$ is an $n \times n$ matrix and let $T({\bf x})=A{\bf x}$. We would like to construct a matrix representation of $T$ relative to a nonstandard basis. The idea is to convert vectors into coordinates relative to the new basis, apply the transformation, and then express the result back in the same coordinate system. 

Let $\mathcal{B}=\{ {\bf b}_1, \cdots {\bf b}_n\}$ be a basis for $\mathbb{R}^n$. Then we have a coordinate map ${\bf x} \mapsto [{\bf x}]_{\mathcal{B}}$. In this case, it is simply the inverse of $P_{\mathcal{B}}$. This produces a new transformation that maps $[{\bf x}]_{\mathcal{B}}$ to $[T({\bf x})]_{\mathcal{B}}$. This new transformation does exactly the same thing as $T$, but in the coordinate system determined by $\mathcal{B}$.

The image below describes what we are doing in the case of $\mathbb{R}^2$. The top part shows the transformation $T$ with its domain and codomain. The bottom half shows the coordinate spaces obtained after applying the coordinate map to both the domain and codomain. We are looking for a transformation between these coordinate spaces that maps $[{\bf x}]_{\mathcal{B}}$ to $[T({\bf x})]_{\mathcal{B}}$.


<center> <img src="data:image/png;base64,  
 " width="40%"> </center>

To find the matrix for this new transformation, we track what happens to the standard basis vectors.
We do this in three steps:

1. Undo the coordinate map. Since, for any $i \leq n$, $[{\bf b}_i]_{\mathcal{B}} = {\bf e}_i$, vector ${\bf e}_i$ goes to ${\bf b}_i$.
2. Apply $T$. This gives us $T({\bf b}_i)$
3. Express the result in $\mathcal{B}$-coordinates. This gives $[T({\bf b}_i)]_{\mathcal{B}}$

So overall, ${\bf e}_i$ goes to $[T({\bf b}_i)]_{\mathcal{B}}$ and the standard matrix for this new transformation is $$M=\begin{bmatrix} [T({\bf b}_1)]_{\mathcal{B}} & [T({\bf b}_2)]_{\mathcal{B}} & \cdots & [T({\bf b}_n)]_{\mathcal{B}}\end{bmatrix}$$


This matrix is called **$\mathcal{B}$-matrix for $T$**. In practice, we usually begin directly with Step 2, since the basis vectors are already known.

    margin: 0.4em 0;
">
<strong>📌 Note:</strong> Let $P$ be the inverse of the coordinate map (i.e. $P=P_{\mathcal{B}}$). Then $A=PMP^{-1}$. So, $A$ and $M$ are similar. This reflects the fact that they represent the same linear transformation in different coordinate systems.
</div>



**<font color="red">Example 4</font>**

Let $A=\begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix}$ and $T({\bf x})=A{\bf x}$. Let $\mathcal{B}=\{{\bf b}_1, {\bf b}_2\}$, where ${\bf b}_1=\begin{bmatrix} 1 \\ 1 \end{bmatrix}, {\bf b}_2= \begin{bmatrix} 0 \\ -1 \end{bmatrix} $ be a basis for $\mathbb{R}^2$. Find the $\mathcal{B}$-matrix for $T$.

**Solution**

First, we find the images of the basis vectors under $T$

$$T({\bf b}_1)=T\left(\begin{bmatrix} 1 \\ 1 \end{bmatrix}\right)=\begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix}\begin{bmatrix} 1 \\ 1 \end{bmatrix}=\begin{bmatrix} 2 \\ 2 \end{bmatrix}$$
$$T({\bf b}_2)=T\left(\begin{bmatrix} 0 \\ -1 \end{bmatrix}\right)=\begin{bmatrix} 1 & 1 \\ 2 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ -1 \end{bmatrix}=\begin{bmatrix} -1 \\ 0 \end{bmatrix}$$

Next, we find the coordinate vectors of the images relative to the basis $\mathcal{B}$. To do this, we need to solve two linear systems:
$$\begin{bmatrix} {\bf b}_1 & {\bf b}_2 \end{bmatrix}{\bf x}=\begin{bmatrix} 2 \\ 2 \end{bmatrix}$$
$$\begin{bmatrix} {\bf b}_1 & {\bf b}_2 \end{bmatrix}{\bf x}=\begin{bmatrix}-1 \\ 0 \end{bmatrix}$$

We can solve both systems simultaneously by augmenting the coefficient matrix with both right-hand sides:
$$\left[
\begin{array}{cc|cc}
1 & 0  & 2 & -1 \\ 1 & -1  & 2 & 0
\end{array}
\right] \sim \cdots \sim
\left[
\begin{array}{cc|cc}
1 & 0  & 2 & -1 \\ 0 & 1  & 0 & -1
\end{array}
\right]$$

The last two columns are the coordinate vectors of the images of the basis vectors. Therefore, the $\mathcal{B}$-matrix for $T$ is

$$M=\begin{bmatrix} 2 & -1 \\0 & -1 \end{bmatrix}$$

---

Suppose $A$ is diagonalizable and let $T({\bf x})=A{\bf x}$. Let $\mathcal{B}$ be an eigenbasis relative to $A$, and $P$ be a matrix whose columns are the vectors in $\mathcal{B}$. Then the $\mathcal{B}$-matrix for $T$ is $P^{-1}AP$. This means that the $\mathcal{B}$-matrix for $T$ is a diagonal matrix $D$ with the eigenvalues along its diagonal. 

---
**<font color="red">Example 5</font>**

Let $A=\begin{bmatrix} 1 & 1 & 2 \\ 1& 4  & 1 \\ 2 & 1 & 1 \end{bmatrix}$ and $T({\bf x})=A{\bf x}$. Find a basis $\mathcal{B}$ such that the $\mathcal{B}$-matrix for $T$ is diagonal, if possible.

**Solution**

For the $\mathcal{B}$-matrix for $T$ to be diagonal, $\mathcal{B}$ must be an eigenbasis relative to $A$. We begin by finding the eigenvalues:

$$| A-\lambda I | = \begin{vmatrix} 1-\lambda & 1 & 2 \\ 1 & 4-\lambda & 1 \\ 2 & 1 & 1-\lambda \end{vmatrix}= -\lambda^3 +6\lambda^2 - 3\lambda -10=-(\lambda-5)(\lambda-2)(\lambda+1)=0$$

So, we have three eigenvalues $\lambda=5$, $\lambda=2$ and $\lambda=-1$. We now know that $A$ is diagonalizable, and hence, a diagonal $\mathcal{B}$-matrix for $T$ exists. Next, find an eigenvector for each eigenvalue:

$$\lambda =5 \rightarrow  \begin{bmatrix} 1 \\ 2 \\ 1 \end{bmatrix}; \quad \lambda =2 \rightarrow  \begin{bmatrix} 1 \\ -1 \\ 1 \end{bmatrix}; \quad \lambda =-1 \rightarrow  \begin{bmatrix} -1 \\0 \\ 1 \end{bmatrix}$$

Hence, an eigenbasis for $A$ is $$\mathcal{B}=\left\{\begin{bmatrix} 1 \\ 2 \\ 1 \end{bmatrix},\begin{bmatrix} 1 \\ -1 \\ 1 \end{bmatrix},\begin{bmatrix} -1 \\0 \\ 1 \end{bmatrix}\right\}$$

Relative to this basis, the $\mathcal{B}$-matrix for $T$ is diagonal: $$\begin{bmatrix} 5 & 0 & 0 \\ 0 & 2 & 0 \\ 0 & 0 & -1\end{bmatrix}$$
