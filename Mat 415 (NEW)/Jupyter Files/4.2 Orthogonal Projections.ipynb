<b>Orthogonal Projection onto a Line</b>
</div></font> 

Orthogonality plays a central role in linear algebra, and many of its applications rely on the idea of **orthogonal projection**. To build some intuition, consider the following two related problems.

**Problem 1.**   
Suppose we have a line $\mathit{L}$ passing through the origin and a vector ${\bf v}$. Imagine light shining perpendicularly onto the line $\mathit{L}$. We want to find the shadow that the vector ${\bf v}$ casts onto the line. More precisely, we want to find the shadow vector along the line $\mathit{L}$ whose magnitude gives the length of the shadow.

**Problem 2.**

Suppose we have a line $\mathit{L}$ passing through the origin and a vector ${\bf v}$. We want to find the shortest distance from ${\bf v}$ to the line $\mathit{L}$.

The figures below illustrate how these two problems are related. The distance from the tip of ${\bf v}$ to the tip of its shadow is exactly the shortest distance from ${\bf v}$ to the line $\mathit{L}$.

This shadow is precisely what we call the **orthogonal projection of ${\bf v}$ onto the line $\mathit{L}$**. 

<center> <img src="data:image/png;base64,  
 " width="60%"> </center>

Let's make it more precise. Let ${\bf u}$ be a nonzero vector lying on the line $\mathit{L}$, and let $\bf s$ denote the shadow vector (that is, the projection). Since ${\bf s}$ lies on the line, it must be a scalar multiple of ${\bf u}$; that is, ${\bf s}=c{\bf u}$ for some scalar $c$. Therefore, determining $c$ allows us to find the projection.

Let ${\bf z}={\bf v}-{\bf s}={\bf v}-c{\bf u}$.  Notice that ${\bf z}$ must be orthogonal to the line $\mathit{L}$, or equivalently, orthogonal to the vector ${\bf u}$. 

<center> <img src="data:image/png;base64,  
 " width="30%"> </center>

Therefore, 
$$({\bf v}-c{\bf u}) \cdot  {\bf u} =0$$
$$ {\bf v}\cdot{\bf u}-c{\bf u} \cdot {\bf u} =0$$
$$c{\bf u} \cdot {\bf u}={\bf v}\cdot{\bf u}$$
$$c=\frac{{\bf v}\cdot{\bf u}}{{\bf u} \cdot {\bf u}}$$

Hence, the projection is given by $${\bf s}=c{\bf u}=\frac{{\bf v}\cdot{\bf u}}{{\bf u} \cdot {\bf u}} {\bf u}$$


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition.</strong> <br> If ${\bf u}$ and ${\bf v}$ are vectors in $\mathbb{R}^n$ and ${\bf u} \neq {\bf 0}$, then the <strong>orthogonal projection of ${\bf v}$ onto the line spanned by ${\bf u}$</strong> is the vector $proj_{\bf u}({\bf v})$ defined by

$$proj_{\bf u}({\bf v})=\frac{{\bf v}\cdot{\bf u}}{{\bf u} \cdot {\bf u}} {\bf u}$$
</div>


What can we say about the vector ${\bf z}$? Since ${\bf z}={\bf v}-proj_{\bf u}({\bf v})$, its norm $||{\bf z}||=||{\bf v}-proj_{\bf u}({\bf v})||$ is exactly the shortest distance from ${\bf v}$ to the line $\mathit{L}$.


This result leads to the following useful theorem.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> If ${\bf u}$ and ${\bf v}$ are vectors in $\mathbb{R}^n$ and ${\bf u} \neq {\bf 0}$, then ${\bf v}$ can be uniquely decomposed into a sum of two orthogonal vectors, ${\bf v}={\bf s}+{\bf z}$, where ${\bf s}$ is a scalar multiple of ${\bf u}$. More precisely, $${\bf s}=proj_{\bf u}({\bf v})$$ and $${\bf z}={\bf v}-proj_{\bf u}({\bf v})$$ 
The vector ${\bf s}$ is called the <strong>vector component of ${\bf v}$ along ${\bf u}$</strong>, and the vector ${\bf z}$ is called the <strong>vector component of ${\bf v}$ orthogonal to ${\bf u}$</strong>.
</div>


---
**<font color="red">Example 1</font>**

Let ${\bf v}=\begin{bmatrix} 1 \\3\\4\end{bmatrix}$ and ${\bf u}=\begin{bmatrix} 2 \\0\\-2\end{bmatrix}$. Find the projection $proj_{\bf u}({\bf v})$ and the distance from ${\bf v}$ to the line spanned by ${\bf u}$.

**Solution.**

Let's first compute the required dot products:

$${\bf v} \cdot {\bf u} = (1)(2)+(3)(0)+(4)(-2)=-6$$
$${\bf u} \cdot {\bf u} = (2)(2)+(0)(0)+(-2)(-2)=8$$

So, the projection is 

$$proj_{\bf u}({\bf v})=\frac{{\bf v}\cdot{\bf u}}{{\bf u} \cdot {\bf u}} {\bf u}=\frac{-6}{8}\begin{bmatrix} 2 \\0\\-2 \end{bmatrix}=\begin{bmatrix}-3/2\\0 \\ 3/2 \end{bmatrix}$$

Next, we find the distance between ${\bf v}$ and the line:

$$||{\bf v}-proj_{\bf u}({\bf v})||=\left|\left|\begin{bmatrix} 1 \\3\\4\end{bmatrix}- \begin{bmatrix}-3/2\\0 \\ 3/2 \end{bmatrix} \right|\right|=\left|\left|\begin{bmatrix}5/2\\3 \\ 5/2 \end{bmatrix} \right|\right|=\sqrt{6.25+9+6.25}=\sqrt{21.5}$$

The projection formula is closely related to the coefficients in the orthogonal basis theorem from the previous chapter. In fact, that theorem expresses every vector as a sum of its orthogonal projections onto the basis vectors. For example, suppose $\mathcal{B}=\{{\bf u}_1, {\bf u}_2\}$ is an orthogonal basis for $\mathbb{R}^2$ and let $\bf v$ be any vector in $\mathbb{R}^2$. Then,  
$${\bf v} = \frac{{\bf v}\cdot {\bf u}_1}{{\bf u}_1 \cdot {\bf u}_1} {\bf u}_1 + \frac{{\bf v}\cdot {\bf u}_2}{{\bf u}_2 \cdot {\bf u}_2} {\bf u}_2$$

Since each coefficient represents an orthogonal projection, we can write this more compactly as

$${\bf v} = proj_{{\bf u}_1}({\bf v})  + proj_{{\bf u}_2}({\bf v}) $$

<b> Orthogonal Projection onto a Subspace</b>
</div></font> 

Let $\mathcal{B}=\{{\bf u}_1, {\bf u}_2\}$ be an orthogonal basis for $\mathbb{R}^2$. Let $W=span({\bf u}_1)$ be a subspace of $\mathbb{R}^2$. Then its orthogonal complement is $W^{\perp}=span({\bf u}_2)$. Therefore, every vector $\bf v$ in $\mathbb{R}^2$ can be decomposed into the sum of two vectors: one lying in the subspace $W$ and the other lying in its orthogonal complement $W^{\perp}$. 

We now generalize this idea to subspaces of $\mathbb{R}^n$. In other words, given a vector ${\bf v}$ in $\mathbb{R}^n$ and a subspace $W$ of $\mathbb{R}^n$, we want to find vectors ${\bf w}_1 \in W$ and ${\bf w}_2 \in W^{\perp}$ such that ${\bf v}={\bf w}_1+{\bf w}_2$.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> Let $W$ be a subspace of $\mathbb{R}^n$ and let $\{{\bf u}_1, \cdots , {\bf u}_k \}$ be an orthogonal basis for $W$. For any vector $\bf v$ in $\mathbb{R}^n$, the <strong> orthogonal projection of $\bf v$ onto $W$ </strong> is defined by
$$ proj_W({\bf v})=\frac{{\bf v} \cdot {\bf u}_1}{{\bf u}_1 \cdot {\bf u}_1}{\bf u}_1+\cdots+\frac{{\bf v} \cdot {\bf u}_k}{{\bf u}_k \cdot {\bf u}_k}{\bf u}_k$$

The vector ${\bf z}={\bf v}-proj_W({\bf v})$ lies in $W^{\perp}$ and is called the <strong> component of $\bf v$ orthogonal to $W$ </strong>.
</div>


Each term in the above sum is itself the projection of $\bf v$ onto the line spanned by one of the basis vectors. Hence,
$$ proj_W({\bf v})=proj_{{\bf u}_1}({\bf v})+\cdots+proj_{{\bf u}_k}({\bf v})$$

The following theorem shows that the orthogonal projection onto a subspace always exists and is unique. Moreover, it does not depend on the choice of orthogonal basis for the subspace.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Let $W$ be a subspace of $\mathbb{R}^n$ and let $\bf v$ be a vector in $\mathbb{R}^n$. Then there exist unique vectors $\bf w \in W$ and ${\bf w}^{\perp} \in W^{\perp}$ such that $${\bf v}={\bf w}+{\bf w}^{\perp}$$
</div>



**Proof.**

We first prove existence.

Let $\{{\bf u}_1, \cdots , {\bf u}_k \}$ be any orthogonal basis for $W$. Let ${\bf w}=proj_W({\bf v})$ and ${\bf w}^{\perp}={\bf v}-{\bf w}$. Then ${\bf w}$ is in $W$ since, by definition, it is a linear combination of vectors in $W$.

To show that ${\bf w}^{\perp}$ is in $W^{\perp}$, it suffices to show that it is orthogonal to every basis vector ${\bf u}_i$. 

$${\bf u}_i \cdot {\bf w}^{\perp}={\bf u}_i \cdot ( {\bf v}-{\bf w} )= {\bf u}_i \cdot  {\bf v}- {\bf u}_i \cdot  {\bf w}$$
$$={\bf u}_i \cdot  {\bf v}- {\bf u}_i \cdot \left(\frac{{\bf u}_1 \cdot {\bf v}}{{\bf u}_1 \cdot {\bf u}_1} {\bf u}_1+ \cdots + \frac{{\bf u}_i \cdot {\bf v}}{{\bf u}_i \cdot {\bf u}_i}{\bf u}_i+ \cdots + \frac{{\bf u}_k \cdot {\bf v}}{{\bf u}_k \cdot {\bf u}_k}{\bf u}_k\right)$$
$$={\bf u}_i \cdot  {\bf v}- \left(\frac{{\bf u}_1 \cdot {\bf v}}{{\bf u}_1 \cdot {\bf u}_1} ( {\bf u}_i \cdot{\bf u}_1)+ \cdots + \frac{{\bf u}_i \cdot {\bf v}}{{\bf u}_i \cdot {\bf u}_i}( {\bf u}_i \cdot {\bf u}_i)+ \cdots + \frac{{\bf u}_k \cdot {\bf v}}{{\bf u}_k \cdot {\bf u}_k}( {\bf u}_i \cdot {\bf u}_k)\right)$$
$$={\bf u}_i \cdot  {\bf v}- \left({\bf 0} + \cdots + \frac{{\bf u}_i \cdot {\bf v}}{{\bf u}_i \cdot {\bf u}_i}( {\bf u}_i \cdot {\bf u}_i)+ \cdots + {\bf 0}\right)$$
$$={\bf u}_i \cdot  {\bf v}-{\bf u}_i \cdot  {\bf v} = 0$$

Therefore, ${\bf w}^{\perp}$ is orthogonal to every basis vector of $W$, and hence belongs to $W^{\perp}$.

Finally, ${\bf w}+{\bf w}^{\perp}={\bf w} + ( {\bf v} - {\bf w}) = {\bf v}$, proving existence.

To prove uniqueness, suppose we have another decomposition ${\bf v}={\bf y}+{\bf y}^{\perp}$, where $\bf y$ is in $W$ and ${\bf y}^{\perp}$ is in $W^{\perp}$. Then $${\bf w}+{\bf w}^{\perp}= {\bf y}+{\bf y}^{\perp}$$
Therefore, 
$${\bf w}- {\bf y}= {\bf y}^{\perp}-{\bf w}^{\perp}$$
However, since $ {\bf w}- {\bf y}$ belongs to $W$ and ${\bf y}^{\perp}-{\bf w}^{\perp}$ belongs to $W^{\perp}$. Since the only common vector $W$ and $W^{\perp}$ have is the zero vector, we must have that
$${\bf w}- {\bf y}= {\bf y}^{\perp}-{\bf w}^{\perp} = {\bf 0}$$
Hence, $${\bf w}= {\bf y} \quad \text{ and } \quad {\bf y}^{\perp}={\bf w}^{\perp}$$

**<font color="red">Example 2</font>**

Let ${\bf u}_1=\begin{bmatrix} 1 \\ -2 \\ 1\end{bmatrix}$, ${\bf u}_2=\begin{bmatrix} -1 \\ 1 \\ 3 \end{bmatrix}$ and ${\bf y}=\begin{bmatrix} 2 \\ 0 \\ 8 \end{bmatrix}$. Let $W=span({\bf u}_1, {\bf u}_2)$. Write ${\bf y}$ as a sum of a vector in $W$ and a vector in $W^{\perp}$.

**Solution.**

First note that $\{{\bf u}_1, {\bf u}_2 \}$ is an orthogonal basis for $W$.
We first find the projection of $\bf y$ onto $W$:

$$proj_W({\bf y})=\frac{{\bf y} \cdot {\bf u}_1}{{\bf u}_1 \cdot {\bf u}_1} {\bf u}_1+\frac{{\bf y} \cdot {\bf u}_2}{{\bf u}_2 \cdot {\bf u}_2} {\bf u}_2 = \frac{10}{6}\begin{bmatrix} 1 \\ -2 \\ 1\end{bmatrix} + \frac{22}{11}\begin{bmatrix} -1 \\ 1 \\ 3 \end{bmatrix}= \begin{bmatrix} 5/3 \\ -10/3 \\ 5/3 \end{bmatrix}+ \begin{bmatrix} -2 \\ 2 \\ 6 \end{bmatrix} = \begin{bmatrix} -1/3 \\ -4/3 \\ 23/3 \end{bmatrix}$$

To obtain the component in $W^{\perp}$, we subtract:

$${\bf y}-proj_W({\bf y}) = \begin{bmatrix} 2 \\ 0 \\ 8 \end{bmatrix} - \begin{bmatrix} -1/3 \\ -4/3 \\ 23/3 \end{bmatrix} = \begin{bmatrix} 7/3 \\ 4/3 \\ 1/3 \end{bmatrix}$$

---

If $\{{\bf u}_1, \cdots , {\bf u}_k \}$ is an orthogonal basis for $W$ and a vector ${\bf v} \in W$, then ${\bf v}$ can be expressed as a linear combination of the basis vectors. In fact, the projection formula is exactly the same as the representation formula for an orthogonal basis from the previous chapter. Since $\bf v$ already belongs to $W$, projecting it onto $W$ leaves it unchanged. Therefore, $proj_W({\bf v})={\bf v}$.

<b>Projection Matrices</b>
</div></font> 

Using an orthonormal basis makes the projection formula much simpler. Suppose  $\{{\bf u}_1, \cdots , {\bf u}_k \}$ is an orthonormal basis for a subspace $W$ of $\mathbb{R}^n$. Then the orthogonal projection of a vector $\bf v$ onto $W$ is given by 
$$proj_W({\bf v})=({\bf v}\cdot {\bf u}_1){\bf u}_1 + \cdots + ({\bf v}\cdot {\bf u}_k){\bf u}_k$$

We can rewrite this expression as a matrix multiplication.
Let  $U=\begin{bmatrix} {\bf u}_1 & \cdots &{\bf u}_k\end{bmatrix}$. Since the columns of $U$ form an orthonormal set, we have $$proj_W({\bf v})=UU^T{\bf v}$$ 

To see why this is true, notice that $proj_W({\bf v})$ is a linear combination of columns of $U$ with coefficients $({\bf v}\cdot {\bf u}_1), \cdots, ({\bf v}\cdot {\bf u}_k)$. So, the projection can be written as a product of $U$ and a vector whose entries are those coefficients:

$$proj_W({\bf v})=U\begin{bmatrix} {\bf v}\cdot {\bf u}_1 \\ \vdots \\ {\bf v}\cdot {\bf u}_k \end{bmatrix}$$

Since ${\bf v} \cdot {\bf u}_i={\bf u}_i^T{\bf v}$, the coefficient vector can be written as $U^T\bf v$, which gives $$proj_W({\bf v})=UU^T{\bf v}$$

<strong>📌 Note:</strong> The matrix $P=UU^T$ is called the <strong> projection matrix</strong> onto the subspace $W$.
</div>

**<font color="red">Example 3</font>**

Let ${\bf u}_1=\begin{bmatrix} 1 \\2 \\ 3 \\ 4 \\5 \\6 \end{bmatrix}$, ${\bf u}_2=\begin{bmatrix} -4 \\-3 \\ 2 \\ 1 \\0\\0 \end{bmatrix}$, ${\bf u}_3=\begin{bmatrix} 0 \\0 \\ -2 \\ 4 \\4 \\-5 \end{bmatrix}$ and ${\bf v}=\begin{bmatrix} 4 \\4 \\ 4 \\ 4 \\4 \\4 \end{bmatrix}$. Using Python, find the projection matrix $P=UU^T$ and the orthogonal projection of $\bf v$ onto the subspace spanned by ${\bf u}_1,{\bf u}_2$ and ${\bf u}_3$.

**Solution.**

Notice that ${\bf u}_1,{\bf u}_2,{\bf u}_3$ form an orthogonal set, but not an orthonormal set. We first place these vectors into the columns of a matrix and then use `la.norm` to normalize all the columns simultaneously. The argument `axis=0` tells `la.norm` to compute the norm along the first axis (down the rows), producing one norm for each column. If we used `axis=1` instead, the function would compute the norm along the second axis (across the columns), producing one norm for each row.

```python
v=np.array([4,4,4,4,4,4])
U_notnorm=np.array([[1,2,3,4,5,6],[-4,-3, 2,1,0,0],[0,0,-2,4,4,-5]]).T

U=U_notnorm/la.norm(U_notnorm, axis=0)
print(U)
```

Now that we have a matrix whose columns form an orthonormal basis, we can find the projection matrix:

```python
P=U@U.T
print(P)
```

The orthogonal projection of $\bf v$ onto the subspace spanned by ${\bf u}_1,{\bf u}_2$ and ${\bf u}_3$ is

```python
proj=P@v
print(proj)
```

The following theorem summarizes some important properties of matrices with orthonormal columns.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Properties of a matrix with orthonormal columns</strong><br> Let $U$ be an $n \times m$ matrix whose columns form an orthonormal set, and let $W=Col(U)$. Then<br>

<strong> 1.</strong> $U^TU{\bf v} = {\bf v}$ for any vector $\bf v$ in $\mathbb{R}^m$ <br>
<strong>2.</strong> $UU^T{\bf u}=proj_W({\bf u})$ for any vector $\bf u$ in $\mathbb{R}^n$<br>
<strong>3.</strong> If $U$ is square, then $UU^T{\bf u}=I{\bf u}={\bf u}$. Therefore, $proj_W({\bf u})={\bf u}$
</div>

The last property also follows directly from the definition of projection. If $U$ is square, then its columns form an orthonormal basis for $\mathbb{R}^n$. Therefore, $W=Col(U)=\mathbb{R}^n$. So, every vector $\bf u$ already belongs to $W$. Consequently, $proj_W({\bf u})={\bf u}$.
