<b>Matrix of the Quadratic Form</b>
</div></font> 

For most of this course, we have dealt with expressions of the form $$c_1x_1+ c_2x_2 +\cdots + c_nx_n$$
If we treat $c_1, \cdots , c_n$ as constants, this expression can be viewed as a function from $\mathbb{R}^n$ to $\mathbb{R}$ whose variables are $x_1, \cdots , x_n$. It is called a **linear form** on $\mathbb{R}^n$. All variables in a linear form appear to the first power, and we do not multiply variables together. 

Similarly, we can define **quadratic forms** on $\mathbb{R}^n$. In a quadratic form, each term contains either a variable raised to the second power or a product of two variables:

$$c_1x_1^2+ c_2x_2^2 + \cdots + c_nx_n^2 + d_1x_1x_2+d_2x_1x_3 + \cdots + d_kx_{n-1}x_n$$

Terms of the form $d_ix_ix_j$ with $i \neq j$ are called cross-terms. Since $x_ix_j=x_jx_i$, it is common to combine these terms.

For example, the general quadratic form on $\mathbb{R}^2$ is $$c_1x_1^2+c_2x_2^2+2c_3x_1x_2$$
while the general quadratic form on $\mathbb{R}^3$ is $$c_1x_1^2+c_2x_2^2+c_3x_3^2+2c_4x_1x_2+2c_5x_1x_3+2c_6x_2x_3$$

Every quadratic form on $\mathbb{R}^n$ can be expressed in the form ${\bf x}^TA{\bf x}$, where $\bf x$ is a vector in $\mathbb{R}^n$ with entries $x_1, \cdots, x_n$ and $A$ is an $n \times n$ symmetric matrix.

---
**<font color="red">Example 1</font>**

Let ${\bf x}=\begin{bmatrix} x_1 \\ x_2\end{bmatrix}$, $A=\begin{bmatrix} 2 & 3 \\ 3 & -4\end{bmatrix}$, $B=\begin{bmatrix} 2 & 0 \\ 0 & -4\end{bmatrix}$. Compute ${\bf x}^TA{\bf x}$, ${\bf x}^TB{\bf x}$, and ${\bf x}^TI{\bf x}$.

**Solution.**

<strong>1.</strong> For $A$: $${\bf x}^TA{\bf x}=\begin{bmatrix} x_1 & x_2 \end{bmatrix}\begin{bmatrix} 2 & 3 \\ 3 & -4\end{bmatrix} \begin{bmatrix} x_1 \\ x_2\end{bmatrix}=\begin{bmatrix} x_1 & x_2 \end{bmatrix}\begin{bmatrix} 2x_1 +3x_2\\ 3x_1-4x_2\end{bmatrix}=2x_1^2+3x_2x_1+3x_1x_2-4x_2^2=2x_1^2+6x_1x_2-4x_2^2$$

<strong>2.</strong> For $B$:  $${\bf x}^TB{\bf x}=\begin{bmatrix} x_1 & x_2 \end{bmatrix}\begin{bmatrix} 2 & 0 \\ 0 & -4\end{bmatrix} \begin{bmatrix} x_1 \\ x_2\end{bmatrix}=\begin{bmatrix} x_1 & x_2 \end{bmatrix}\begin{bmatrix} 2x_1 \\ -4x_2\end{bmatrix}=2x_1^2-4x_2^2$$

<strong>3.</strong> For the identity matrix: $${\bf x}^TI{\bf x}={\bf x}^T{\bf x}=\begin{bmatrix} x_1 & x_2 \end{bmatrix} \begin{bmatrix} x_1 \\ x_2\end{bmatrix}=x_1^2+x_2^2 = ||{\bf x}||^2$$

---

From these examples, we can make the following observations:

<strong> 1.</strong> If $A$ is diagonal, then the quadratic form ${\bf x}^TA{\bf x}$ has no cross-terms.<br>
<strong> 2.</strong> If $A$ has a nonzero diagonal entry $a_{ii}$, then the quadratic form has a term $a_{ii}x_i^2$.<br>
<strong> 3.</strong> If $A$ has a nonzero off-diagonal entry $a_{ij}$, then the quadratic form has a cross-term $2a_{ij}x_ix_j$.

Based on these observations, we can rewrite any quadratic form in the form ${\bf x}^TA{\bf x}$.

---
**<font color="red">Example 2</font>**

Rewrite the quadratic form $x_1^2+3x_2^2-5x_3^2+6x_1x_2-12x_2x_3$ in the form ${\bf x}^TA{\bf x}$.

**Solution.**

The matrix $A$ is a $3 \times 3$ matrix, since the quadratic form is defined on $\mathbb{R}^3$. The squared terms $x_1^2+3x_2^2-5x_3^2$ give the diagonal entries: $1, 3, -5$. 

The cross-terms produce off-diagonal entries. Recall that each cross-term coefficient arises from two symmetric off-diagonal entries of $A$. Therefore, the term $6x_1x_2$ corresponds to $\require{color}
\newcommand{\hlo}[1]{\bbox[orange, 2pt]{#1}}\hlo{3}$ in both the $(1,2)$ and $(2,1)$ positions.

Similarly, $-12x_2x_3$ corresponds to $\require{color}
\newcommand{\hldt}[1]{\bbox[DarkTurquoise, 2pt]{#1}}\hldt{-6}$ in both the $(2,3)$ and $(3,2)$ positions.

Since there is no $x_1x_3$ term, the $(1,3)$ and $(3,1)$ entries are $\newcommand{\hlyg}[1]{\bbox[YellowGreen, 2pt]{#1}}\hlyg{0}$.

Hence, 
$$\require{color}
\newcommand{\hlo}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hldt}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
\newcommand{\hlyg}[1]{\bbox[YellowGreen, 2pt]{#1}}
A=\begin{bmatrix} 1 & \hlo{3} & \hlyg{0} \\ \hlo{3} & 3 & \hldt{-6} \\ \hlyg{0} & \hldt{-6} & -5 \end{bmatrix}$$

It is easy to verify that $${\bf x}^TA{\bf x}=\begin{bmatrix} x_1 & x_2 & x_3\end{bmatrix}\begin{bmatrix} 1 & 3 & 0 \\ 3 & 3 & -6 \\ 0& -6 & -5 \end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix} =x_1^2+3x_2^2-5x_3^2+6x_1x_2-12x_2x_3$$

<strong>🏛️ Definition</strong><br> Let $A$ be an $n \times n$ symmetric matrix. The function $Q({\bf x})={\bf x}^TA{\bf x}$ is called the <strong> quadratic form associated with $A$ </strong>, and $A$ is called the <strong>matrix of the quadratic form $Q$.</strong>
</div>




**<font color="red">Example 3</font>**

Let $A=\begin{bmatrix} 1 & 3 & 0 \\ 3 & 3 & -6 \\ 0& -6 & -5 \end{bmatrix}$ and let $Q({\bf x})={\bf x}^TA{\bf x}$ be the quadratic form associated with $A$. Evaluate $Q({\bf x})$ for ${\bf x}=\begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}$, ${\bf x}=\begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}$, and ${\bf x}=\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$.

**Solution.**

The matrix $A$ is the same as in Example 2. Therefore, $Q(x_1, x_2, x_3)=x_1^2+3x_2^2-5x_3^2+6x_1x_2-12x_2x_3$. Then,

<strong> 1.</strong>  $Q(1,2,3)=1^2+3(2)^2-5(3)^2+6(1)(2)-12(2)(3)=1+12-45+12-72=-92$
   
<strong> 2.</strong>  $Q(0, 1, 0)=0^2+3(1)^2-5(0)^2+6(0)(1)-12(1)(0)=3$

<strong> 3.</strong>  $Q(0, 0, 0)=0^2+3(0)^2-5(0)^2+6(0)(0)-12(0)(0)=0$

The last two evaluations illustrate two general properties of quadratic forms: $Q({\bf 0})=0$ and $Q({\bf e}_i)=c_i$, where $c_i$ is a coefficient of $x_i^2$.

<b>Change of Variables</b>
</div></font>

It is generally easier to work with quadratic forms that have no cross-terms. In this case, the matrix of the quadratic form is diagonal. 

Since the matrix of a quadratic form is symmetric, the Spectral Theorem tells us that it can be orthogonally diagonalized. This allows us to perform a change of variables that eliminates all cross-terms. 

Suppose $Q({\bf x})={\bf x}^TA{\bf x}$, where $A$ is symmetric. A **change of variables** is a substitution of the form ${\bf x}=P{\bf y}$, where $P$ is an invertible matrix and $\bf y$ is a new variable vector. Geometrically, this can be viewed as expressing ${\bf x}$ in a new coordinate system. The vector $\bf y$ contains the coordinates of $\bf x$ with respect to the basis formed by the columns of $P$. Applying this change of variables to the quadratic form $Q$ gives:

$${\bf x}^TA{\bf x}=(P{\bf y})^TA(P{\bf y})={\bf y}^T(P^TAP){\bf y}$$

Thus, the change of variables transforms the quadratic form into a new quadratic form whose matrix is $P^TAP$. Since $A$ is symmetric, the Spectral Theorem guarantees that there is an orthogonal matrix $P$ such that $P^TAP=D$, where $D$ is diagonal. Hence,
$${\bf x}^TA{\bf x}={\bf y}^TD{\bf y}$$
From the previous chapter, we know that the columns of $P$ are orthonormal eigenvectors of $A$, while the diagonal entries of $D$ are the corresponding eigenvalues. When $P$ is chosen in this way, we call ${\bf x}=P{\bf y}$ an **orthogonal change of variables**.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> Let $A$ be an $n\times n$ symmetric matrix. Then there exists an orthogonal change of variables ${\bf x}=P{\bf y}$ that transforms the quadratic form ${\bf x}^TA{\bf x}$ into a quadratic form ${\bf y}^TD{\bf y}$ with no cross-terms. Specifically, if $P$ orthogonally diagonalizes $A$, then $${\bf x}^TA{\bf x}={\bf y}^TD{\bf y}=\lambda_1y_1^2+\cdots + \lambda_ny_n^2,$$ where $\lambda_1, \cdots, \lambda_n$ are the eigenvalues of $A$ corresponding to the eigenvectors that form the columns of $P$, and $y_1, \cdots, y_n$ are the entries of the new variable vector ${\bf y}$.
</div>



The directions determined by the columns of $P$ in the above theorem are called **principal axes** of the quadratic form ${\bf x}^TA{\bf x}$.

---
**<font color="red">Example 4</font>**

Find an orthogonal change of variables that eliminates the cross-term in the quadratic form $Q(x_1, x_2)=x_1^2-8x_1x_2-5x_2^2$

**Solution.**

The matrix associated with the quadratic form $Q$ is $$A=\begin{bmatrix} 1 & -4 \\ -4 & -5 \end{bmatrix}$$

Its eigenvalues are $\lambda=3$ and $\lambda=-7$, with corresponding normalized eigenvectors $\begin{bmatrix} 2/\sqrt{5} \\ -1/\sqrt{5} \end{bmatrix}$ and $\begin{bmatrix} 1/\sqrt{5} \\ 2/\sqrt{5} \end{bmatrix}$. These eigenvectors form an orthonormal basis of $\mathbb{R}^2$. Therefore, 

$$P=\begin{bmatrix} 2/\sqrt{5} & 1/\sqrt{5} \\ -1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix} \text{ and } D=\begin{bmatrix} 3 & 0 \\ 0 & -7\end{bmatrix}$$

Hence, an orthogonal change of variables ${\bf x}=P{\bf y}$ eliminates the cross-terms:

$${\bf x}^TA{\bf x}={\bf y}^T(P^TAP){\bf y}={\bf y}^TD{\bf y}=\begin{bmatrix}y_1 & y_2\end{bmatrix} \begin{bmatrix} 3 & 0 \\ 0 & -7\end{bmatrix}\begin{bmatrix}y_1 \\ y_2\end{bmatrix}=3y_1^2-7y_2^2$$

Thus, the original quadratic form $x_1^2-8x_1x_2-5x_2^2$ becomes $3y_1^2-7y_2^2$ in the new coordinate system.

It is worth noting that a change of variables does not change the values attained by the quadratic form. It only changes the coordinates used to represent those values. For example, let ${\bf x}= \begin{bmatrix} 1 \\ 3 \end{bmatrix}$. Then $${\bf y}=P^{-1}{\bf x}=P^T{\bf x}=\begin{bmatrix} 2/\sqrt{5} & -1/\sqrt{5} \\  1/\sqrt{5} & 2/\sqrt{5} \end{bmatrix}\begin{bmatrix} 1 \\ 3 \end{bmatrix}=\begin{bmatrix} -1/\sqrt{5} \\ 7/\sqrt{5} \end{bmatrix}$$

Evaluating the original quadratic form gives

$$x_1^2-8x_1x_2-5x_2^2=1^2-8(1)(3)-5(3^2)=1-24-45=-68$$

Using the new coordinates gives the same result: 

$$3y_1^2-7y_2^2 = 3\left(\frac{-1}{\sqrt{5}}\right)^2-7\left(\frac{7}{\sqrt{5}}\right)^2=\frac{3}{5}-\frac{343}{5}=-\frac{340}{5}=-68$$

Thus, the change of variables has not changed the quadratic form itself; it has simply expressed the same function in a coordinate system aligned with the eigenvector directions, where all cross-terms disappear.

<b>Geometric Point of View</b>
</div></font> 

For ease of visualization, we will mostly focus on quadratic forms on $\mathbb{R}^2$ in this section. However, the results we discuss remain valid for quadratic forms in $\mathbb{R}^n$.

One question that often arises while working with quadratic forms is: what geometric shape is represented by the equation ${\bf x}^TA{\bf x}=c$ for some constant $c$? 

If $A$ is a $2 \times 2$ symmetric matrix, then this equation can represent an ellipse, a hyperbola, two intersecting lines, two parallel lines, a single line, a single point, or no points at all. If $A$ is diagonal, we say that the graph is in **standard position**; otherwise, the graph may be rotated relative to the coordinate axes. The orthogonal matrix $P$ from the diagonalization of $A$ determines the rotation or reflection needed to express the graph in standard position.

**<font color="red">Example 5</font>**

Consider the quadratic form $Q(x_1, x_2)=x_1^2-8x_1x_2-5x_2^2$ from Example 4. What is the graph of $Q(x_1, x_2)=1$?

**Solution.**

From Example 4, we know that the principal axes of $Q$ are given by the vectors $${\bf v}_1=\begin{bmatrix} 2/\sqrt{5} \\ -1/\sqrt{5} \end{bmatrix} \text{ and } {\bf v}_2=\begin{bmatrix} 1/\sqrt{5} \\ 2/\sqrt{5} \end{bmatrix}$$

We also found that the orthogonal change of variables ${\bf x}=P{\bf y}$ transforms the quadratic form into $3y_1^2-7y_2^2$. Therefore, in the new coordinate system, the equation $Q(x_1, x_2)=1$ becomes $3y_1^2-7y_2^2=1$. Taking the reciprocals of the coefficients and expressing them as squares, we can rewrite the equation $3y_1^2-7y_2^2=1$ in the standard form of a hyperbola: $$\frac{y_1^2}{(1/\sqrt{3})^2}-\frac{y_2^2}{(1/\sqrt{7})^2}=1$$


In the original $x_1x_2$-coordinate system, the hyperbola is rotated relative to the coordinate axes. Its principal axes, the $y_1$- and $y_2$-axes, are aligned with the directions of ${\bf v}_1$ and ${\bf v}_2$, respectively:

<b>Classification of Quadratic Forms</b>
</div></font> 

Quadratic forms can be classified according to the signs of the values they take.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> A quadratic form $Q({\bf x})={\bf x}^TA{\bf x}$ is

<strong>1.</strong> <strong>positive definite</strong> if $Q({\bf x})>0$ for every ${\bf x} \neq {\bf 0}$ <br>
<strong>2.</strong> <strong>positive semidefinite</strong> if $Q({\bf x}) \geq 0$ for every ${\bf x}\in \mathbb{R}^n$ <br>
<strong>3.</strong> <strong>negative definite</strong> if $Q({\bf x})<0$ for every ${\bf x} \neq {\bf 0}$ <br>
<strong>4.</strong> <strong>negative semidefinite</strong> if $Q({\bf x})\leq 0$ for every ${\bf x}\in \mathbb{R}^n$ <br>
<strong>5.</strong> <strong>indefinite</strong> if $Q({\bf x})$ takes on both positive and negative values.
</div>



For example, the four surfaces shown above represent positive definite, negative definite, indefinite, and positive semidefinite quadratic forms, respectively.

We can determine the type of a quadratic form by examining the eigenvalues of its corresponding symmetric matrix.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Let $A$ be an $n \times n$ symmetric matrix. Then the quadratic form $Q({\bf x})={\bf x}^TA{\bf x}$ is

<strong>1.</strong> positive definite if and only if all eigenvalues of $A$ are positive. <br>
<strong>2.</strong> positive semidefinite if and only if all eigenvalues of $A$ are nonnegative. <br>
<strong>3.</strong> negative definite if and only if all eigenvalues of $A$ are negative. <br>
<strong>4.</strong> negative semidefinite if and only if all eigenvalues of $A$ are nonpositive. <br>
<strong>5.</strong> indefinite if and only if some eigenvalues of $A$ are positive and some are negative. <br>
</div>



<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

    border-left: 4px solid #2c3e50;

    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
<strong>📌 Note:</strong>  We also use this terminology to classify the matrix $A$ itself. For instance, if the quadratic form $Q({\bf x})={\bf x}^TA{\bf x}$ is positive definite, we say that the symmetric matrix $A$ is positive definite.
</div>


**<font color="red">Example 6</font>**

Classify the quadratic form $Q(x_1, x_2, x_3)=4x_1^2+4x_2^2+5x_3^2-6x_1x_2-4x_1x_3+4x_2x_3$.

**Solution.**

The corresponding symmetric matrix is 

$$\begin{bmatrix} 4 & -3 & -2 \\ -3 & 4 & 2 \\ -2 & 2 & 5 \end{bmatrix}$$

This matrix has three eigenvalues: $1$, $3$, and $9$. Since all three eigenvalues are positive, the quadratic form is positive definite.

---

In the following chapters, we will frequently encounter matrices of the form $A^TA$. Such matrices are always positive semidefinite. Consequently, all eigenvalues of $A^TA$ are nonnegative.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Let $A$ be an $m \times n$ matrix. Then $A^TA$ is positive semidefinite.
</div>

**Proof.**

Since $$(A^TA)^T=A^T(A^T)^T=A^TA,$$
the matrix $A^TA$ is symmetric.


Now consider the quadratic form ${\bf x}^TA^TA{\bf x}$, where ${\bf x} \in \mathbb{R}^n$. We have
$${\bf x}^TA^TA{\bf x} = (A{\bf x})^T(A{\bf x})=(A{\bf x}) \cdot (A{\bf x}) = ||A{\bf x}||^2 \geq 0$$

Therefore, $A^TA$ is positive semidefinite.
