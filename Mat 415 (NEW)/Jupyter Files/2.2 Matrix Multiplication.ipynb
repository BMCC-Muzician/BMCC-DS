2.2 Matrix Multiplication
</div></font>

We can define matrix multiplication in a couple of ways. One way is to view it as the composition of matrix transformations. Another is to view it as multiplying a matrix by multiple vectors simultaneously (just as matrix–vector multiplication can be viewed as a collection of dot products). Regardless of which perspective we use, we obtain the same multiplication rule. We will start with the first approach.

<b>Matrix Product as a Composition</b>
</div></font> 

Suppose we have two matrix transformations $T:\mathbb{R}^k \rightarrow \mathbb{R}^m$ given by $T({\bf x})=A{\bf x}$, and $S:\mathbb{R}^n \rightarrow \mathbb{R}^k$ given by $S({\bf x})=B{\bf x}$. Based on their domains and codomains, matrix $A$ is an $m\times k$ matrix and matrix $B$ is a $k \times n$ matrix. Consider their composition $$(T \circ S) ({\bf x})=T(S({\bf x}))=T(B{\bf x})=A(B{\bf x})=(AB){\bf x}$$

Recall that the composition of two linear transformations is again a linear transformation (see Chapter 2.1 Exercise 10). Therefore, the composition $(T \circ S) ({\bf x})$ must also be a linear transformation from $\mathbb{R}^n$ to $\mathbb{R}^m$, and hence a matrix transformation. Our goal is to find the matrix $(AB)$ such that $(AB){\bf x}=A(B{\bf x})$. The composition matrix $(AB)$ defines a transformation from $\mathbb{R}^n$ to $\mathbb{R}^m$ and therefore it must be an $m \times n$ matrix. 

Note that for this composition to make sense, the domain of $T$ must be the same as the codomain of $S$. This means that the number of columns of $A$ must equal to the number of rows of $B$. (Recall the similar restriction on matrix-vector multiplication).

Suppose $B=\begin{bmatrix} {\bf b}_1 & \cdots & {\bf b}_n\end{bmatrix}$ and ${\bf x}=\begin{bmatrix} x_1 \\ \vdots \\ x_n\end{bmatrix}$, then
$$B{\bf x}=x_1{\bf b}_1+\cdots+x_n{\bf b}_n$$

Applying $A$ to both sides, we get:

$$A(B{\bf x})=A(x_1{\bf b}_1+\cdots+x_n{\bf b}_n)=x_1A{\bf b}_1+\cdots+x_nA{\bf b}_n$$

We can rewrite the last part as a matrix-vector multiplication:

$$x_1A{\bf b}_1+\cdots+x_nA{\bf b}_n=\begin{bmatrix} A{\bf b}_1 & \cdots & A{\bf b}_n\end{bmatrix}\begin{bmatrix} x_1 \\ \vdots \\ x_n\end{bmatrix}=\begin{bmatrix} A{\bf b}_1 & \cdots & A{\bf b}_n\end{bmatrix}{\bf x}$$

Therefore, $$ AB=\begin{bmatrix} A{\bf b}_1 & \cdots & A{\bf b}_n\end{bmatrix}$$

Notice that this product is defined by multiplying $A$ by every column of $B$.

---
**<font color="red">Example 1</font>**

Let $A=\begin{bmatrix} 1 & 2 \\ 3 & 4 \\ -1 & 1\end{bmatrix}$, $B=\begin{bmatrix} 3 & 1 \\ -2 & 5 \end{bmatrix}$. Find $AB$.

**Solution**

$$A{\bf b}_1=\begin{bmatrix} 1 & 2 \\ 3 & 4 \\ -1 & 1\end{bmatrix}\begin{bmatrix} 3 \\ -2 \end{bmatrix}=\begin{bmatrix} -1 \\ 1 \\ -5\end{bmatrix} \quad\text{and} \quad A{\bf b}_2=\begin{bmatrix} 1 & 2 \\ 3 & 4 \\ -1 & 1\end{bmatrix}\begin{bmatrix} 1 \\ 5 \end{bmatrix}=\begin{bmatrix} 11 \\ 23 \\ 4\end{bmatrix}$$
And so, 
$$AB=\begin{bmatrix} -1 & 11 \\ 1 & 23 \\ -5 & 4\end{bmatrix}$$

<b>Matrix Product Using Row–Column Method</b>
</div></font> 

Another way to compute matrix products is the row–column method that we used in matrix-vector multiplication. Suppose ${\bf a}_i^T$ is $i$-th row of $A$ and ${\bf b}_j$ is $j$-th column of $B$. The $(i, j)$-entry of $AB$ is equal to a dot product of ${\bf a}_i$ and ${\bf b}_j$, where ${\bf a}_i$ is the column vector whose transpose is the $i$-th row of $A$. To see why this is true, notice that the $j$-th column of $AB$, by the definition, is equal to $A{\bf b}_j$. Moreover, $i$-th entry of $A{\bf b}_j$ can be calculated using row–column method as  ${\bf a}_i^T{\bf b}_j = ({\bf a}_i^T)^T \cdot {\bf b}_j ={\bf a}_i \cdot {\bf b}_j $.

This method is an efficient way of performing matrix multiplication by hand.


---
**<font color="red">Example 2</font>**

Let $A=\begin{bmatrix} 1 & 2 \\ 3 & -1 \end{bmatrix}$, $B=\begin{bmatrix} -3 & 0 \\ 2 & -2 \end{bmatrix}$. Find $AB$ using row–column method.

**Solution**

We perform a dot product of every row of $A$ with every column of $B$ and write the result in the appropriate place in the resulting matrix. Here is an example of the dot product of the first row of $A$ and second column of $B$:

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
\left[
\begin{array}\fbox{\begin{array}{cc}
\hl{1} & \hl{2} \\ 3 & -1
\end{array}}\end{array}
\right]
\left[
\begin{array}\fbox{\begin{array}{cc}
-3 & \hl{0} \\ 2 & \hl{-2}
\end{array}}\end{array}
\right]=
\left[
\begin{array}\fbox{\begin{array}{cc}
* & \hl{1(0)+ 2(-2)} \\ * & *
\end{array}}\end{array}
\right]=
\left[
\begin{array}\fbox{\begin{array}{cc}
* & \hl{-4} \\ * & *
\end{array}}\end{array}
\right]
$$

We use the same technique for other entries of $AB$:
 $$AB=\begin{bmatrix} 1 & 2 \\ 3 & -1 \end{bmatrix}\begin{bmatrix} -3 & 0 \\ 2 & -2 \end{bmatrix}= \begin{bmatrix} 1(-3)+2(2) & 1(0)+ 2(-2) \\ 3(-3)-1(2) & 3(0)-1(-2) \end{bmatrix}=\begin{bmatrix} 1 & -4 \\ -11 & 2 \end{bmatrix}$$

<strong>🏛️ Properties of Matrix Multiplication </strong> <br> Let  $A$, $B$ and $C$ be matrices (with appropriate dimensions so that all products and sums are well-defined), and let $c$ be a scalar. Then<br>

<strong>1. </strong> $A(BC)=(AB)C$  <br>
<strong>2. </strong> $A(B+C)=AB+AC$  <br>
<strong>3. </strong> $(A+B)C=AC+BC$  <br>
<strong>4. </strong> $c(AB)=(cA)B=A(cB)$  <br>
<strong>5. </strong> $A=I_mA=AI_n$, if $A$ is an $m \times n$ matrix. <br>

</div>


Notice that commutativity is missing from the properties above. Unlike multiplication of real numbers, matrix multiplication is not commutative in general.

<strong>3. </strong> No zero product rule: If $AB=0$, then it does not necessarily follow that $A=0$ or $B=0$.

We can use power notation to express repeated products of matrices. Given a square matrix $A$ and a nonnegative integer $n$, $A^n$ denotes the product of $A$ with itself $n$ times: 
$$A^n=A\cdots A$$

For $n=0$, we define $A^0=I$, where $I$ is the identity matrix of the same size as $A$. To understand why this definition makes sense, consider a vector ${\bf x}$ as a starting point. We can think of applying $A$ as taking one step. Then $A^2{\bf x}$ corresponds to taking two steps, and $A^n{\bf x}$ corresponds to taking $n$ steps starting from ${\bf x}$. However, $A^0{\bf x}$ should correspond to taking no steps, which means that $A^0{\bf x}={\bf x}=I{\bf x}$

Matrix powers satisfy the familiar exponent rules:

<strong>1.</strong> $A^nA^m=A^{n+m}$<br>
<strong>2.</strong> $(A^n)^m=A^{nm}$

<b>Transpose of a Matrix</b>
</div></font> 

In Chapter 1.1, we introduced the transpose operation for vectors. Recall that transposing a vector exchanges its rows and columns. We can apply the same operation to matrices as well.

---
**<font color="red">Example 3</font>**

Let $A=\begin{bmatrix} 1 & 2 \\ 3 & 4 \\ -1 & 1\end{bmatrix}$. Find the transpose $A^T$.

**Solution**

The first column of $A$ will become the first row of $A^T$, and the second column of $A$ will become the second row of $A^T$:

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}
\newcommand{\hly}[1]{\bbox[DarkTurquoise, 2pt]{#1}}
A=\left[
\begin{array}\fbox{\begin{array}{cc}
\hl{1} & \hly{2} \\ \hl{3} & \hly{4} \\ \hl{-1} & \hly{1}
\end{array}}\end{array}
\right]
\quad{\large\Longrightarrow}\quad
A^T=\left[
\begin{array}\fbox{\begin{array}{cc}
\hl{1} & \hl{3}& \hl{-1} \\ \hly{2}  & \hly{4} & \hly{1}  
\end{array}}\end{array}
\right]
$$

<strong>🏛️ Properties of Transpose </strong> <br> Let  $A$ and $B$ be matrices (with appropriate sizes to make sum or product to be well defined) and let $c$ be a scalar. Then<br>

<strong>1. </strong> $(A^T)^T=A$  <br>
<strong>2. </strong> $(A+B)^T=A^T+B^T$  <br>
<strong>3. </strong> $(cA)^T=c(A^T)$  <br>
<strong>4. </strong> $(AB)^T=B^TA^T$ 

</div>


Note the last property: the transpose of a product is equal to the product of the transposes in reverse order.

In the next example, we will see how matrix multiplication can be used to efficiently evaluate a matrix transformation at many points simultaneously. We will generate 1,000 points and visualize how the given transformation moves them. 

---
**<font color="red">Example 5</font>**


Let $A=\begin{bmatrix} 2 & 0 \\ 1 & -1\end{bmatrix}$.
We use `def` to define the transformation $T({\bf x})=A{\bf x}$

```python
def T(x):
    A=np.array([[2,0],[1,-1]])
    return A@x
```

The next cell creates a $1000 \times 2$ array of random points in $\mathbb{R}^2$. Each **row** of `X` represents a vector in $\mathbb{R}^2$.

```python
mn = scp.stats.multivariate_normal(mean=[0,0], cov=[[1,0],[0,1]])
X = mn.rvs(size=1000)
print(X)
```

Let's plot all the points in `X` and all their images under the transformation $T$ that we will store in `Y`. The `np.arctan2(X[:,1], X[:,0])` part calculates the angle between each vector and the positive x-axis. We will use this angle to color the vectors so that we can track them after applying the transformation.

```python
fig, ax = plt.subplots(1, 2, sharey=True, sharex=True, figsize=(10, 5))
ax[0].set_ylim([-6,6])
ax[0].set_xlim([-6,6])

Y=T(X.T).T  

ax[0].scatter(X[:,0], X[:,1], s = 1, c=np.arctan2(X[:,1], X[:,0]), cmap='hsv')
ax[1].scatter(Y[:,0], Y[:,1], s = 1, c=np.arctan2(X[:,1], X[:,0]), cmap='hsv')
plt.show()
```

Notice that we used the transpose operation twice in `Y=T(X.T).T` to calculate the images. Think about why this is necessary.

From the images, we can observe the following:

<strong>1.</strong> In the left image, if we start in a blue region and move clockwise, the colors transition from blue to green and then to yellow. In the right image, the same sequence of colors appears when moving counterclockwise. This indicates that the transformation reverses orientation. In other words, it includes a reflection (flip).<br>
<strong>2.</strong> We can also observe that the transformation changes the shape and scale of the point cloud. This suggests that the transformation includes stretching and/or shearing.
