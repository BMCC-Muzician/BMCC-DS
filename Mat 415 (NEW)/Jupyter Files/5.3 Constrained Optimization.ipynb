5.3 Constrained Optimization
</div></font>

In this chapter, we will consider the problem of finding the maximum and minimum values of a quadratic form $Q({\bf x})$ over a specified set of vectors ${\bf x}$, which typically will be restricted to a set of unit vectors. This requirement can be stated in several equivalent ways: $$||{\bf x}||=1\text{, } \quad||{\bf x}||^2=1 \text{, }\quad{\bf x}^T{\bf x}=1\text{, }\quad x_1^2+x_2^2+ \cdots +x_n^2=1$$

To help with visualization, consider the figure below. Here, we have a quadratic form on $\mathbb{R}^2$, visualized as a surface in three-dimensional space. The restriction $||{\bf x}||=1$ corresponds to the unit circle in the $xy$-plane. The problem of finding the extrema is therefore equivalent to finding the highest and lowest points on the intersection of the surface with the cylinder determined by the unit circle.

When the quadratic form has no cross-terms, this optimization problem has a simple solution.

---
**<font color="red">Example 1</font>**

Find the maximum and minimum values of $Q({\bf x})=8x_1^2+4x_2^2+2x_3^2$ subject to the constraint $||{\bf x}||=1$.

**Solution.**

Since $x_2^2$ and $x_3^2$ are nonnegative, $$4x_2^2 \leq 8x_2^2 \quad \text{and}\quad 2x_3^2 \leq 8x_3^2$$ Therefore,

$$Q({\bf x})=8x_1^2+4x_2^2+2x_3^2 \leq 8x_1^2+8x_2^2+8x_3^2 = 8(x_1^2+x_2^2+x_3^2) = 8,$$
since $x_1^2+x_2^2+x_3^2=1$. 

So, $Q({\bf x})$ must be less than or equal to $8$. Moreover, when ${\bf x}=(1,0,0)$, we have $Q({\bf x})=8$.
Therefore, $8$ is the maximum value of $Q({\bf x})$ subject to $||{\bf x}||=1$.

Similarly, by observing that $8x_1^2 \geq 2x_1^2$ and $4x_2^2 \geq 2x_2^2$, we can show that $2$ is the minimum value of $Q({\bf x})$ subject to $||{\bf x}||=1$. 

---

Notice that since there are no cross-terms, the eigenvalues of the matrix associated with $Q$ in Example 1 are $2$, $4$, and $8$. The extrema of $Q$ given that $||{\bf x}||=1$ are precisely the largest and smallest eigenvalues. The maximum value, $8$, is achieved at $\pm {\bf e}_1$, while the minimum value, $2$, is achieved at $\pm {\bf e}_3$. These are unit eigenvectors corresponding to the largest and smallest eigenvalues, respectively. The same idea holds for any quadratic form.

<strong>🏛️ Theorem </strong><br> Let $A$ be a symmetric $n \times n$ matrix whose eigenvalues are $\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_n$. Then <br>
<strong>1.</strong> The quadratic form ${\bf x}^TA{\bf x}$, subject to the constraint $||{\bf x}||=1$, has a maximum value of $\lambda_1$ and a minimum value of $\lambda_n$;<br>
<strong>2.</strong> The maximum value is achieved at any unit eigenvector corresponding to $\lambda_1$;<br>
<strong>3.</strong> The minimum value is achieved at any unit eigenvector corresponding to $\lambda_n$.
</div>



**Proof.**

Since $A$ is symmetric, it can be orthogonally diagonalized as $A=PDP^T$, where $P=\begin{bmatrix} {\bf u}_1 & \cdots & {\bf u}_n\end{bmatrix}$ is an orthogonal matrix whose columns ${\bf u}_1,  \cdots,{\bf u}_n$ are unit eigenvectors corresponding to $\lambda_1 , \cdots,  \lambda_n$, respectively. Applying the orthogonal change of variables ${\bf x}=P{\bf y}$ gives $${\bf x}^TA{\bf x}={\bf y}^TD{\bf y}$$

Since $P$ is orthogonal, $||{\bf x}||=||P{\bf y}||=||{\bf y}||$. Therefore, as ${\bf x}$ ranges over all unit vectors, so does ${\bf y}$. Hence, ${\bf x}^TA{\bf x}$ and ${\bf y}^TD{\bf y}$ assume the same set of values.

The matrix $D$ is diagonal, with diagonal entries $\lambda_1, \cdots, \lambda_n$. Therefore, by the same argument as in Example 1, the maximum value of ${\bf y}^TD{\bf y}$, subject to $||{\bf y}||=1$, is $\lambda_1$, while the minimum value is $\lambda_n$. The maximum is achieved at ${\bf y}={\bf e}_1$, and the minimum is achieved at ${\bf y}={\bf e}_n$.

Since ${\bf x}^TA{\bf x}$ and ${\bf y}^TD{\bf y}$ attain the same set of values, $\lambda_1$ and $\lambda_n$ are also the maximum and minimum values  of ${\bf x}^TA{\bf x}$, respectively.

Finally, if ${\bf u}$ is any unit eigenvector corresponding to an eigenvalue $\lambda$, then $A{\bf u}=\lambda{\bf u}$, and hence

$$
{\bf u}^TA{\bf u}=\lambda{\bf u}^T{\bf u}=\lambda
$$

Therefore, every unit eigenvector corresponding to $\lambda_1$ attains the maximum value, and every unit eigenvector corresponding to $\lambda_n$ attains the minimum value.

---
**<font color="red">Example 2</font>**

Let $A=\begin{bmatrix} 5 & 1 & 2 \\ 1 & 5 & 2 \\2 & 2 & 4\end{bmatrix}$. Find the maximum value of the quadratic form ${\bf x}^TA{\bf x}$ subject to the constraint $||{\bf x}||=1$, and find a unit vector at which this maximum is achieved.

**Solution.**

The maximum value of ${\bf x}^TA{\bf x}$ subject to the constraint $||{\bf x}||=1$ is equal to the largest eigenvalue of $A$. The eigenvalues of $A$ are $2$, $4$, and $8$. Therefore, the maximum value is $8$. 

This maximum is achieved at any unit eigenvector corresponding to the eigenvalue $8$. Solving $(A-8I){\bf x}={\bf 0}$ gives the eigenvector $\begin{bmatrix} 1 \\ 1 \\ 1 \end{bmatrix}$. Normalizing this vector gives the unit eigenvector $\begin{bmatrix} 1/\sqrt{3} \\ 1/\sqrt{3} \\ 1/\sqrt{3} \end{bmatrix}$. Its negative is also a unit eigenvector corresponding to $8$. Therefore, the maximum is achieved at either of the following unit vectors: $$\begin{bmatrix} 1/\sqrt{3} \\ 1/\sqrt{3} \\ 1/\sqrt{3} \end{bmatrix}\quad \text{ and }\quad\begin{bmatrix} -1/\sqrt{3} \\ -1/\sqrt{3} \\ -1/\sqrt{3} \end{bmatrix}$$

---

Let's add another constraint to our optimization problem. Suppose ${\bf v}_1$ is a unit eigenvector corresponding to the largest eigenvalue of a symmetric matrix $A$. We now want to find the maximum value of ${\bf x}^TA{\bf x}$ subject to the constraints $||{\bf x}||=1$ and ${\bf x}^T{\bf v}_1=0$. The second constraint ${\bf x}^T{\bf v}_1=0$ means we consider only vectors orthogonal to ${\bf v}_1$. Let's see how this works using the quadratic form from Example 1.

---
**<font color="red">Example 3</font>**

Find the maximum value of $Q({\bf x})=8x_1^2+4x_2^2+2x_3^2$, subject to the constraints $||{\bf x}||=1$ and ${\bf x}^T{\bf e}_1=0$. Note that ${\bf e}_1$ is a unit eigenvector corresponding to the largest eigenvalue $8$ of the matrix of the quadratic form $Q$.

**Solution.**

The condition ${\bf x}^T{\bf e}_1=0$ means that $x_1=0$. Therefore, we need to maximize $4x_2^2+2x_3^2$ subject to $x_2^2+x_3^2=1$. By applying a technique similar to the one used in Example 1, we conclude that the maximum value is $4$, which is the second-largest eigenvalue of the matrix of the quadratic form $Q$. This value is achieved at ${\bf e}_2$, which is a unit eigenvector corresponding to the eigenvalue $4$.

<strong>🏛️ Theorem </strong><br> Let $A$ be a symmetric matrix. Let $\lambda_1$ be the largest eigenvalue of $A$, and let ${\bf v}_1$ be a unit eigenvector corresponding to $\lambda_1$. Then the maximum value of ${\bf x}^TA{\bf x}$ subject to the constraints $||{\bf x}||=1$ and ${\bf x}^T{\bf v}_1=0$ is the second-largest eigenvalue $\lambda_2$. This maximum is achieved at any unit eigenvector corresponding to $\lambda_2$ that is orthogonal to ${\bf v}_1$.
</div>



We can extend this result to all eigenvalues of a quadratic form:

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

    border-left: 4px solid #2c3e50;

    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> Let $A$ be an $n \times n$ symmetric matrix. Let $\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_n$ be the eigenvalues of $A$, with corresponding orthonormal eigenvectors ${\bf v}_1, {\bf v}_2, \cdots, {\bf v}_n$, respectively. Then for $i=2,\cdots, n$, the maximum value of ${\bf x}^TA{\bf x}$ subject to the constraints 


$$||{\bf x}||=1, \quad {\bf x}^T{\bf v}_1=0, \quad\cdots, \quad{\bf x}^T{\bf v}_{i-1}=0$$

is the eigenvalue $\lambda_i$. This maximum is achieved at ${\bf x}={\bf v}_i$.
</div>


In the following example, we will use Python to find some of these extrema.

---
**<font color="red">Example 4</font>**

Let $A=\begin{bmatrix}8&-2&-4&0\\-2&8&0&-4\\-4&0&8&-2\\0&-4&-2&8\end{bmatrix}$.

**a.** Find the maximum value of the quadratic form ${\bf x}^TA{\bf x}$ subject to the constraint $||{\bf x}||=1$.

**b.** Find the maximum value of the quadratic form ${\bf x}^TA{\bf x}$ subject to the constraints $||{\bf x}||=1$ and ${\bf x}^T{\bf v}_1=0$, where ${\bf v}_1$ is a unit eigenvector at which the maximum in part **a** is achieved.

**Solution.**

We can solve both parts of this problem simultaneously. For part **a**, we need the largest eigenvalue, and for part **b**, we need the second-largest eigenvalue.

When working with symmetric matrices, we can use `np.linalg.eigh()` to compute their eigenvalues and eigenvectors. Unlike `np.linalg.eig()`, which works with general square matrices, `np.linalg.eigh()` is specifically designed for symmetric (or Hermitian) matrices and is generally more efficient and numerically stable. This can be particularly beneficial when working with large matrices. It also returns eigenvalues in ascending order.

```python
A=np.array([[8, -2, -4, 0],[-2, 8, 0, -4],[-4, 0, 8, -2],[0, -4, -2, 8]])

eigenvalues, eigenvectors = np.linalg.eigh(A)
print(f'Eigenvalues \n {eigenvalues}')
print(f'Eigenvectors \n {eigenvectors}')
```

Thus, the largest eigenvalue is $14$, and the maximum is achieved at ${\bf x}=\begin{bmatrix} 0.5 \\ -0.5 \\ -0.5 \\ 0.5 \end{bmatrix}$.

The second-largest eigenvalue is $10$, and the corresponding constrained maximum is achieved at ${\bf x}=\begin{bmatrix} -0.5 \\ -0.5 \\ 0.5 \\ 0.5 \end{bmatrix}$.
