<b>Matrix Inverse</b>
</div></font> 

In mathematics, the concept of an inverse is very important as it is fundamental to solving equations. Even solving a simple equation like $2x+1=5$ requires the use of inverses. First, we use the additive inverse of $1$, which is $-1$, to obtain:
$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}2x+1\hl{-1}=5\hl{-1}$$
$$2x=4$$
Next, we use the multiplicative inverse of $2$, which is $1/2$, to obtain:

$$\require{color}\require{bbox}
\newcommand{\hl}[1]{\bbox[orange, 2pt]{#1}}\hl{\frac{1}{2}}(2x)=\hl{\frac{1}{2}}(4)$$
$$x=2$$

However, if we were given the equation $0x=3$, it would have no solution because $0$ does not have a multiplicative inverse.

Our goal is to solve matrix equation $A{\bf x}={\bf b}$. Using logic similar to that of the multiplicative inverse above, we would like to know whether there exists a matrix $B$ such that $BA=I$. This would allow us to multiply both sides of the equation by $B$, obtaining $$BA{\bf x}=B{\bf b}$$
$${\bf x}=B{\bf b}$$

To fully mirror the scalar case, we require both $AB=I$ and $BA=I$. This means that $A$ and $B$ must be square matrices of the same dimensions.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition.</strong> <br> A square matrix $A$ is <strong>invertible</strong> if there exists a square matrix $B$ such that 
$$AB=I\quad\text{ and }\quad BA=I$$
The matrix $B$ is called the <strong>inverse</strong> of $A$ and is denoted by $A^{-1}$
If no such matrix exists, then $A$ is called <strong>non-invertible</strong> or <strong>singular</strong>.
</div>


In general, computing inverses can be computationally intensive. However, we can find the inverse of a $2\times 2$ matrix using a simple formula:


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem.</strong> <br> Let $A=\begin{bmatrix} a & b \\ c & d\end{bmatrix}$. If $ad-bc\neq 0$, then $A$ is invertible and $$A^{-1}=\frac{1}{ad-bc}\begin{bmatrix} d & -b \\-c& a\end{bmatrix}$$ If $ad-bc = 0$, then $A$ is singular.
</div>


The quantity $ad-bc$ is called the determinant of the $2 \times 2$ matrix $A$, denoted by $det(A)=ad-bc$. We will discuss determinants in the future.

---
**<font color="red">Example 1</font>**

Let  $A=\begin{bmatrix} 1 & 4 \\ -2 & -9\end{bmatrix}$. 

**1.** Find $A^{-1}$.<br>
**2.** Verify that $AA^{-1}=I$ and $A^{-1}A=I$

**Solution**

**1.** First, we calculate the determinant $$det(A)=(1)(-9)-(4)(-2)=-1$$

Since the determinant is nonzero, the inverse of $A$ exists:
$$A^{-1}=\frac{1}{-1}\begin{bmatrix} -9 & -4 \\ 2 & 1\end{bmatrix}=\begin{bmatrix} 9 & 4 \\ -2 & -1\end{bmatrix}$$

**2.**

$$AA^{-1}=\begin{bmatrix} 1 & 4 \\ -2 & -9\end{bmatrix}\begin{bmatrix} 9 & 4 \\ -2 & -1\end{bmatrix}=\begin{bmatrix} 1(9)+4(-2) & 1(4)+4(-1) \\ -2(9)+(-9)(-2) & -2(4)+(-9)(-1)\end{bmatrix}=\begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix}$$
$$A^{-1}A=\begin{bmatrix} 9 & 4 \\ -2 & -1\end{bmatrix}\begin{bmatrix} 1 & 4 \\ -2 & -9\end{bmatrix}=\begin{bmatrix} 9(1)+4(-2) & 9(4)+4(-9) \\ -2(1)+(-1)(-2) & -2(4)+(-1)(-9)\end{bmatrix}=\begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix}$$

---
<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem.</strong><br> If $A$ is invertible, then for every ${\bf b}$, the equation $A{\bf x}={\bf b}$ has a unique solution given by ${\bf x}=A^{-1}{\bf b}$
</div>

---
**<font color="red">Example 2</font>**

Let  $A=\begin{bmatrix} 2 & 4 \\ 2 & 3\end{bmatrix}$ and ${\bf b}=\begin{bmatrix} 2 \\1\end{bmatrix}$. Solve $A{\bf x}={\bf b}$ using the inverse of $A$.

**Solution**

First, we calculate the inverse of $A$:
$$A^{-1}=\frac{1}{(2)(3)-(4)(2)}\begin{bmatrix} 3 & -4 \\ -2 & 2 \end{bmatrix}=\frac{1}{-2}\begin{bmatrix} 3 & -4 \\ -2 & 2 \end{bmatrix}=\begin{bmatrix} -3/2 & 2 \\ 1 & -1 \end{bmatrix}$$

We can now solve for $\bf x$:
$${\bf x}=A^{-1}{\bf b}=\begin{bmatrix} -3/2 & 2 \\ 1 & -1 \end{bmatrix}\begin{bmatrix} 2 \\1\end{bmatrix} = \begin{bmatrix} -1 \\1\end{bmatrix}$$

">
<strong>🏛️ Properties of Inverses. </strong> <br> Let $A$ and $B$ be invertible matrices. Then<br>

<strong>1. </strong> $(A^{-1})^{-1}=A$  <br>
<strong>2. </strong> $(AB)^{-1}=B^{-1}A^{-1}$  <br>
<strong>3. </strong> $(A^T)^{-1}=(A^{-1})^T$  <br>

</div>


Notice that, similarly to the transpose operation, taking the inverse of a product reverses the order of the factors.

The inverse we have defined is a multiplicative inverse. Therefore, the exponent $-1$, which denotes the inverse, follows the usual rules of exponents, just as $2^{-1}=1/2$ is the multiplicative inverse of $2$.

If $A$ is an invertible square matrix and $n$ is a positive integer, then we can define $A^{-n}=(A^{-1})^n$. Thus, we can use any integer as an exponent, and the usual rules of exponents still hold.

<b>Algorithm for Finding Inverse</b>
</div></font> 

To develop a general method for finding inverses, we introduce elementary matrices.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition.</strong> <br> An <strong>elementary matrix</strong> is a matrix obtained by performing a single elementary row operation on the identity matrix.
</div>


If $E$ is an elementary matrix, then $EA$ performs the corresponding elementary row operation on $A$.

---
**<font color="red">Example 3</font>**

Let $A=\begin{bmatrix} 1 & 2 \\ 3& 4 \\ 5 &6\end{bmatrix}$ and $E=\begin{bmatrix} 1 & 0 & 0\\ 0& 0& 1\\0&1&0\end{bmatrix}$. 

Notice that $E$ is an elementary matrix obtained by swapping rows 2 and 3 of the identity matrix. Let's compute $EA$:

$$EA=\begin{bmatrix} 1 & 0 & 0\\ 0& 0& 1\\0&1&0\end{bmatrix}\begin{bmatrix} 1 & 2 \\ 3& 4 \\ 5 &6\end{bmatrix}=\begin{bmatrix} 1 & 2  \\ 5 &6\\ 3& 4\end{bmatrix}$$

Notice that the result is simply $A$ with rows 2 and 3 swapped. 

---

Elementary matrices allow us to represent row operations as matrix multiplications. Moreover, since every elementary row operation is reversible, elementary matrices must be invertible. The inverse of an elementary matrix is also an elementary matrix.



<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem.</strong> <br> A matrix $A$ is invertible if and only if it is row equivalent to the identity matrix.
</div>


**Proof**

This is an "if and only" statement, so we have to prove both directions.

**1.** First, we prove that if $A$ is invertible, then it is row equivalent to the identity matrix.

Suppose $A$ is invertible. Then the equation $A{\bf x}={\bf b}$ has a unique solution for every vector $\bf b$. Therefore, $A$ has a pivot in every row. Since $A$ is a square matrix, its reduced row echelon form must be the identity matrix.

**2.** Next, we prove that if $A$ is row equivalent to the identity matrix, then $A$ is invertible.

If $A$ is row equivalent to the identity matrix $I$, then there exists elementary matrices $E_1, \cdots , E_k$ such that $$E_k\cdots E_1A=I$$

Let $B=E_k\cdots E_1$. Since elementary matrices are invertible and the product of invertible matrices is invertible, $B$ is also invertible. Therefore,
$$BA=I$$
$$B^{-1}BA=B^{-1}I$$
$$A=B^{-1}$$

Hence, $A$ is invertible. Moreover, $$A^{-1}=B=E_k\cdots E_1$$

---

Notice that the proof above also shows that if $A$ is invertible, then the same row operations that transform $A$ into the identity matrix transform the identity matrix into $A^{-1}$:
$$E_k\cdots E_1A=I$$
$$E_k\cdots E_1I=A^{-1}$$

We can use this to create an algorithm for finding inverses.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Algorithm for finding $A^{-1}$.</strong> <br> Row-reduce the augmented matrix $\left[
\begin{array}{c|c}
A & I
\end{array}
\right]$. 

If $A$ is invertible, then $$\left[
\begin{array}{c|c}
A & I
\end{array}
\right]\sim\left[
\begin{array}{c|c}
I & A^{-1}
\end{array}
\right]$$
</div>



**<font color="red">Example 4</font>**

Find the inverse of the matrix $$A=\begin{bmatrix} 2 & 1 & 1 \\ 2 & 2 & 0 \\ 1 & 0 & 2 \end{bmatrix}$$

**Solution**

We augment $A$ with the identity matrix and then row reduce it:

$$\left[
\begin{array}\fbox{\begin{array}{c|c}
A& I
\end{array}}\end{array}
\right]=
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
2 & 1 & 1 & 1 & 0& 0 \\ 2 & 2 & 0 & 0 & 1 & 0\\  1 & 0 & 2 & 0 & 0 & 1
\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_1 \leftrightarrow R_3}}{\sim} 
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 2 & 0 & 0 & 1 \\ 2 & 2 & 0 & 0 & 1 & 0   \\ 2 & 1 & 1 & 1 & 0& 0
\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}\substack{R_2 + (-2)R_1\\ R_3 + (-2)R_1} }  {\sim} 
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 2 & 0 & 0 & 1 \\ 0 & 2 & -4 & 0 & 1 & -2   \\ 0 & 1 & -3 & 1 & 0& -2
\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{R_2 \leftrightarrow R_3}}{\sim} 
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 2 & 0 & 0 & 1 \\ 0 & 1 & -3 & 1 & 0& -2  \\ 0 & 2 & -4 & 0 & 1 & -2
\end{array}}\end{array}
\right]
$$

$$\stackrel{\color{HotPink}{R_3 +(-2) R_2}}{\sim}
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 2 & 0 & 0 & 1 \\ 0 & 1 & -3 & 1 & 0& -2  \\ 0 & 0 & 2 & -2 & 1 & 2
\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}{(1/2) R_3}}{\sim} 
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 2 & 0 & 0 & 1 \\ 0 & 1 & -3 & 1 & 0& -2  \\ 0 & 0 & 1 & -1 & 1/2 & 1
\end{array}}\end{array}
\right]
\stackrel{\color{HotPink}\substack{R_2 + (3)R_3\\ R_1 + (-2)R_3} }  {\sim} 
\left[
\begin{array}\fbox{\begin{array}{ccc|ccc}
1 & 0 & 0 & 2 & -1 & -1 \\ 0 & 1 & 0 & -2 & 3/2 & -1/2  \\ 0 & 0 & 1 & -1 & 1/2 & 1
\end{array}}\end{array}
\right]
=\left[
\begin{array}\fbox{\begin{array}{c|c}
I&A^{-1}
\end{array}}\end{array}
\right]
$$

Since the left part is the identity matrix, on the right we have the inverse of $A$:

$$A^{-1}=\begin{bmatrix} 2 & -1 & -1 \\ -2 & 3/2 & -1/2 \\ -1 & 1/2 & 1 \end{bmatrix}$$

To compute the inverse of a matrix, we can use the SciPy library. Let's solve Example 4 using Python:


```python
A=np.array([[2,1,1],[2,2,0],[1,0,2]])
print ( la.inv(A) )
```

In practice, we rarely compute matrix inverses directly. Finding an inverse is often inefficient and computationally expensive. As we progress through linear algebra, we will discover methods for avoiding explicit matrix inversion whenever possible, as well as techniques for making computations more efficient.

For now, we can compare the efficiency of solving a simple matrix equation using a dedicated solver with that of computing the solution using the inverse.

**<font color="red">Example 5</font>**

Let $A=\begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ -1 & -2 & 0 \end{bmatrix}$ and ${\bf b}=\begin{bmatrix} 1 \\ 1 \\1 \end{bmatrix}$. Solve $A{\bf x}={\bf b}$ using NumPy solver and using the inverse od $A$. Time each method.

**Solution**

```python
A=np.array([[1,2,3],[4,5,6],[-1, -2, 0]])
b=np.array([1,1,1])
```

```python
%%timeit -n 1000 -r 100 
np.linalg.solve(A,b)
```

```python
%%timeit -n 1000 -r 100 
la.inv(A)@b
```

Although the exact timings will vary across machines and platforms, you should generally observe that the dedicated solver is faster.

<b>Inverse Matrix Theorem</b>
</div></font> 

The following theorem is one of the fundamental results of linear algebra. In fact, as we progress through the course, we will add more statements to this theorem. Its main purpose is to demonstrate the interconnectedness of many concepts that we have already learned and will encounter in the future. The theorem contains several equivalent statements. This means that if one of them is true, then all of them are true; and if one of them is false, then all of them are false. Take your time to go through the statements one by one and make sure you understand why they are equivalent.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Inverse Matrix Theorem.  </strong> <br>Let $A$ be a square $n \times n$ matrix. Then the following statements are equivalent: <br>
<strong>1. </strong>  $A$ is invertible <br>
<strong>2. </strong> $A{\bf x}={\bf b}$ has a unique solution for every vector ${\bf b}$ in $\mathbb{R}^n$ <br>
<strong>3. </strong> $A{\bf x}={\bf 0}$ has only the trivial solution<br>
<strong>4. </strong> $A$ is row equivalent to the identity matrix<br>
<strong>5. </strong> $A$ has $n$ pivots<br>
<strong>6. </strong> The columns of $A$ form linearly independent set.<br>
<strong>7. </strong> The columns of $A$ span $\mathbb{R}^n$<br>
<strong>8. </strong> The columns of $A$ form a basis for $\mathbb{R}^n$<br>
<strong>9. </strong> $rank(A)=n$<br>
<strong>10. </strong> $dim(Col(A))=n$<br>
<strong>11. </strong> $dim(Nul(A))=0$<br>
<strong>12. </strong> The transformation $T({\bf x})=A{\bf x}$ is one-to-one<br>
<strong>13. </strong> The transformation $T({\bf x})=A{\bf x}$ is onto<br>
<strong>14. </strong> There exists matrix $B$ such that $AB=I$<br>
<strong>15. </strong> There exists matrix $C$ such that $CA=I$
</div>

The last two statements in the theorem tell us that if $A$ and $B$ are square matrices and $AB=I$, then both matrices are invertible and are inverses of each other. In other words, we do not have check whether $BA=I$.
