<b>Orthogonal Sets</b>
</div></font> 

Recall that two vectors $\bf u$ and $\bf v$ are orthogonal if ${\bf u}\cdot {\bf v}={\bf 0}$.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> A set of vectors $\{ {\bf v}_1, \cdots, {\bf v}_k \}$ in $\mathbb{R}^n$ is called an <strong>orthogonal set </strong> if every pair of distinct vectors in the set is orthogonal. In other words, ${\bf v}_i \cdot {\bf v}_j = 0$ whenever $i \neq j$ for $i,j = 1, \cdots, k$
</div>



**<font color="red">Example 1</font>**

The standard basis $\{ {\bf e}_1, \cdots {\bf e}_n \}$ of $\mathbb{R}^n$, as well as any subset of the standard basis, forms an orthogonal set.

---

**<font color="red">Example 2</font>**

Let ${\bf v}_1=\begin{bmatrix} 1 \\ 2 \\ 2 \end{bmatrix}$, ${\bf v}_2=\begin{bmatrix} 2 \\ -1 \\ 0 \end{bmatrix}$ and ${\bf v}_3=\begin{bmatrix} 2 \\ 4 \\-5 \end{bmatrix}$. Show that $S=\{ {\bf v}_1, {\bf v}_2, {\bf v}_3 \}$ is an orthogonal set.

**Solution.**

Compute the pairwise dot products:

$${\bf v}_1 \cdot {\bf v}_2= (1)(2)+(2)(-1)+(2)(0)=0$$
$${\bf v}_1 \cdot {\bf v}_3= (1)(2)+(2)(4)+(2)(-5)=0$$
$${\bf v}_2 \cdot {\bf v}_3=(2)(2)+(-1)(4)+(0)(-5)=0$$

Since all pairs of distinct vectors are orthogonal, $S$ is an orthogonal set.

---

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> If $S=\{ {\bf v}_1, \cdots {\bf v}_k \}$ is an orthogonal set of non-zero vectors in $\mathbb{R}^n$, then $S$ is a linearly independent set. Consequently, $S$ is a basis for $span(S)$.
</div>



**Proof.**

Suppose $c_1, \cdots c_k$ are scalars such that $c_1{\bf v}_1+ \cdots + c_k{\bf v}_k={\bf 0}$. Then for any $i=1, \cdots k$ we have

$$ ( c_1{\bf v}_1+ \cdots +c_k{\bf v}_k ) \cdot {\bf v}_i = {\bf 0} \cdot {\bf v}_i=0$$

On the other hand,

$$ ( c_1{\bf v}_1+ \cdots + c_k{\bf v}_k ) \cdot {\bf v}_i = c_1({\bf v}_1 \cdot {\bf v}_i) +\cdots + c_k({\bf v}_k \cdot {\bf v}_i)= c_i ({\bf v}_i \cdot {\bf v}_i)$$

Therefore, $$c_i ({\bf v}_i \cdot {\bf v}_i)=0$$

Since ${\bf v}_i$ is nonzero, we must have that $c_i=0$. This holds for any $i=1, \cdots k$. Therefore, $S$ is a linearly independent set.

---

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> An <strong>orthogonal basis </strong> for a subspace $W$ of $\mathbb{R}^n$ is a basis for $W$ that is also an orthogonal set.
</div>



<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Let $\{ {\bf v}_1, \cdots ,{\bf v}_k \}$ be an orthogonal basis for a subspace $W$ of $\mathbb{R}^n$, and let ${\bf y}$ be any vector in $W$. Then the unique weights $c_1, \cdots , c_k $ such that $${\bf y}=c_1{\bf v}_1+ \cdots +c_k{\bf v}_k $$ are given by $$c_i=\frac{{\bf y}\cdot {\bf v}_i}{{\bf v}_i\cdot {\bf v}_i } \quad\text { for } i=1, \cdots, k$$
</div>



**Proof.**

For any $i=1, \cdots k$, we have

$${\bf y}\cdot {\bf v}_i=(c_1{\bf v}_1+ \cdots +c_k{\bf v}_k)\cdot {\bf v}_i = c_i ({\bf v}_i \cdot {\bf v}_i)$$

Since ${\bf v}_i \cdot {\bf v}_i \neq 0$, we can solve for $c_i$:

$$c_i=\frac{{\bf y}\cdot {\bf v}_i}{{\bf v}_i\cdot {\bf v}_i }$$

---

**<font color="red">Example 3</font>**

Let ${\bf v}_1= \begin{bmatrix} 1 \\ 1 \\1 \end{bmatrix}$, ${\bf v}_2=\begin{bmatrix} 2 \\ -1 \\-1 \end{bmatrix}$, ${\bf v}_3=\begin{bmatrix} 0 \\ 3 \\ -3 \end{bmatrix}$ and  ${\bf y}=\begin{bmatrix} 6 \\ 2 \\-4 \end{bmatrix}$. Verify that $\mathcal{B}=\left\{{\bf v}_1, {\bf v}_2, {\bf v}_3  \right\}$ is an orthogonal basis for $\mathbb{R}^3$, and find the coordinate vector $[{\bf y}]_{\mathcal{B}}$.

**Solution.**

First, verify that $\mathcal{B}$ is an orthogonal set:

$${\bf v}_1 \cdot {\bf v}_2= (1)(2)+(1)(-1)+(1)(-1)=0$$
$${\bf v}_1 \cdot {\bf v}_3= (1)(0)+(1)(3)+(1)(-3)=0$$
$${\bf v}_2 \cdot {\bf v}_3=(2)(0)+(-1)(3)+(-1)(-3)=0$$

Now we compute the weights:

$$c_1=\frac{{\bf y}\cdot {\bf v}_1}{{\bf v}_1\cdot {\bf v}_1 }=\frac{(6)(1)+(2)(1)+(-4)(1)}{1+1+1}=\frac{4}{3}$$
$$c_2=\frac{{\bf y}\cdot {\bf v}_2}{{\bf v}_2\cdot {\bf v}_2 }=\frac{(6)(2)+(2)(-1)+(-4)(-1)}{4+1+1}=\frac{14}{6}=\frac{7}{3}$$
$$c_3=\frac{{\bf y}\cdot {\bf v}_3}{{\bf v}_3\cdot {\bf v}_3 }=\frac{(6)(0)+(2)(3)+(-4)(-3)}{0+9+9}=\frac{18}{18}=1$$

Therefore,

$${\bf y}=\frac{4}{3}{\bf v}_1+\frac{7}{3}{\bf v}_2+1  {\bf v}_3$$

And so, the coordinate vector is $$[{\bf y}]_{\mathcal{B}}=\begin{bmatrix} \frac{4}{3}\\\frac{7}{3}\\1\end{bmatrix}$$

When the basis is orthogonal, the coordinates can be computed directly using dot products, making the process much simpler.
</div>

<b>Orthonormal Sets</b>
</div></font> 




<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;

">
<strong>🏛️ Definition</strong><br> A set of vectors in $\mathbb{R}^n$ is called an <strong> orthonormal set </strong> if it is an orthogonal set of unit vectors.
    
An <strong> orthonormal basis</strong> for a subspace $W$ of $\mathbb{R}^n$ is a basis for $W$ that is an orthonormal set.
</div>



**<font color="red">Example 4</font>**

The standard basis $\{ {\bf e}_1, \cdots {\bf e}_n \}$ of $\mathbb{R}^n$ is an orthonormal basis.

---

**<font color="red">Example 5</font>**

Let ${\bf v}_1= \begin{bmatrix} 1/3 \\ 2/3 \\-2/3 \end{bmatrix}$, ${\bf v}_2=\begin{bmatrix} 0 \\ 1/\sqrt{2} \\1/\sqrt{2} \end{bmatrix}$, ${\bf v}_3=\begin{bmatrix} -4/\sqrt{18} \\ 1/\sqrt{18} \\ -1/\sqrt{18} \end{bmatrix}$. Show that $\mathcal{B}=\{ {\bf v}_1, {\bf v}_2, {\bf v}_3 \}$ is an orthonormal basis for $\mathbb{R}^3$.

**Solution.**

We first check whether $S$ is an orthogonal set:

$${\bf v}_1 \cdot {\bf v}_2= 0+\frac{2}{3\sqrt{2}}-\frac{2}{3\sqrt{2}}=0$$
$${\bf v}_1 \cdot {\bf v}_3= -\frac{4}{3\sqrt{18}}+\frac{2}{3\sqrt{18}}+\frac{2}{3\sqrt{18}}=0$$
$${\bf v}_2 \cdot {\bf v}_3=0+\frac{1}{\sqrt{36}}-\frac{1}{\sqrt{36}}=0$$

Next, we check that all three vectors have norm equal to $1$:

$${\bf v}_1 \cdot {\bf v}_1= \frac{1}{9}+\frac{4}{9}+\frac{4}{9}=1$$
$${\bf v}_2 \cdot {\bf v}_2= 0+\frac{1}{2}+\frac{1}{2}=1$$
$${\bf v}_3 \cdot {\bf v}_3=\frac{16}{18}+\frac{1}{18}+\frac{1}{18}=1$$

Thus, $\{ {\bf v}_1, {\bf v}_2, {\bf v}_3 \}$ is an orthonormal set. Since every orthogonal set of nonzero vectors is linearly independent,  $\mathcal{B}=\{ {\bf v}_1, {\bf v}_2, {\bf v}_3 \}$ is an orthonormal basis for $\mathbb{R}^3$.

---

Normalizing an orthogonal set preserves orthogonality. Therefore, if we divide each nonzero vector in an orthogonal set by its norm, the resulting set is orthonormal.

Although orthonormal vectors often involve square roots and can be more cumbersome to work with by hand, orthonormal bases are extremely useful in applications and numerical computations.

One reason is that the coordinate formula from the previous theorem becomes even simpler when we use an orthonormal basis. For an orthonormal basis, the coordinates of a vector are simply its dot products with the basis vectors. This makes coordinate computations especially efficient and is one of the main reasons orthonormal bases are so useful in linear algebra, data science, and scientific computing.

<strong>🏛️ Theorem </strong><br> Let $\{ {\bf v}_1, \cdots ,{\bf v}_k \}$ be an orthonormal basis for a subspace $W$ of $\mathbb{R}^n$, and let ${\bf y}$ be any vector in $W$. Then the unique weights $c_1, \cdots , c_k $  such that $${\bf y}=c_1{\bf v}_1+ \cdots +c_k{\bf v}_k $$ are given by $$c_i={\bf y}\cdot {\bf v}_i \quad\text {  for } i=1, \cdots, k$$
</div>

<b>Orthogonal Matrices</b>
</div></font> 




Matrices whose columns form an orthonormal set arise frequently in applications because they have many useful properties.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong><br> An $m\times n$ matrix $U$ has orthonormal columns if and only if $U^TU=I$
</div>



**Proof.**

Recall that the $(i,j)$-entry of the product $U^TU$ is the dot product of the $i$-th column of $U$ and $j$-th column of $U$.
There are two cases:

1. If $i=j$, then the entry is a dot product of a column with itself. Since each column has unit norm, we have ${\bf u}_i\cdot {\bf u}_i=||{\bf u}_i||^2=1$. 
2. If $i\neq j$, then the entry is a dot product of two distinct columns, but since columns are orthogonal, we have ${\bf u}_i\cdot {\bf u}_j=0$

Therefore, $U^TU$ has ones along main diagonal and zeros everywhere else. Hence $U^TU=I$.

Conversely, if $U^TU=I$, then the diagonal entries show that each column has norm 1, while the off-diagonal entries show that distinct columns are orthogonal. Hence, the columns of $U$ form an orthonormal set.

---

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Let $U$ be an $m\times n$ matrix with orthonormal columns and let $\bf x$ and $\bf y$ be vectors in $\mathbb{R}^n$. Then<br>
<strong>1.</strong> $||U{\bf x}||=||{\bf x}||$<br>
<strong>2.</strong> $(U{\bf x})\cdot (U{\bf y})={\bf x}\cdot {\bf y}$
</div>



**Proof.**

1. Since the columns of $U$ are orthonormal, we have $U^TU=I$. Then
 $$||U{\bf x}||^2=(U{\bf x})\cdot (U{\bf x}) =(U{\bf x})^TU{\bf x}={\bf x}^TU^TU{\bf x}={\bf x}^TI{\bf x}=||{\bf x}||^2$$

Taking square root of both sides gives $$||U{\bf x}||=||{\bf x}||$$

2. We use similar reasoning for the second part:
   $$(U{\bf x})\cdot (U{\bf y})=(U{\bf x})^T(U{\bf y})={\bf x}^TU^TU{\bf y}={\bf x}^T{\bf y}={\bf x}\cdot {\bf y}$$

---

The two properties above show that the transformation $T({\bf x})=U{\bf x}$ preserves the lengths of vectors and the angles between them.

<strong>🏛️ Definition</strong><br> A square $n \times n$ matrix $U$ with orthonormal columns is called an <strong> orthogonal matrix </strong>
</div>

</div> 

The next theorem is probably the most important property of orthogonal matrices.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> Matrix $U$ is orthogonal if and only if $U^{-1}=U^T$
</div>



The importance of this theorem is hard to overstate. Computing the inverse of a matrix can be computationally expensive, so we try to avoid it whenever possible. However, finding the transpose of a matrix is one of the simplest matrix operations.

---
**<font color="red">Example 6</font>**

Show that $A=\begin{bmatrix} \sin(x) & \cos(x) \\ -\cos(x) & \sin(x) \end{bmatrix}$ is orthogonal and find its inverse.

**Solution.**

We compute $$A^TA=\begin{bmatrix} \sin(x) & -\cos(x) \\ \cos(x) & \sin(x) \end{bmatrix}\begin{bmatrix} \sin(x) & \cos(x) \\ -\cos(x) & \sin(x) \end{bmatrix}=\begin{bmatrix} \sin^2(x)+\cos^2(x) & \sin(x)\cos(x)-\cos(x)\sin(x) \\ \cos(x)\sin(x)-\sin(x)\cos(x) & \cos^2(x)+\sin^2(x) \end{bmatrix}=\begin{bmatrix} 1 & 0 \\0 & 1 \end{bmatrix}=I$$

Since $A^TA=I$ and $A$ is a square matrix, $A$ is an orthogonal matrix. Therefore,

$$A^{-1}=A^T=\begin{bmatrix} \sin(x) & -\cos(x) \\ \cos(x) & \sin(x) \end{bmatrix}$$
