<b>Gram-Schmidt Process</b>
</div></font> 

In the previous two chapters, we showed that orthogonal and orthonormal bases are very convenient. However, we often begin with a basis that is not orthogonal. It would therefore be useful to have a way to convert a non-orthogonal basis into an orthogonal basis for the same subspace. The **Gram-Schmidt process** does exactly that. The idea is quite simple: we start with a non-orthogonal basis and use projections to orthogonalize it one vector at a time.

---
**<font color="red">Example 1</font>**

Let ${\bf x}_1 =\begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}$ and ${\bf x}_2=\begin{bmatrix} -1 \\ 1 \\ 2 \end{bmatrix}$. Let $\mathcal{B}=\left\{{\bf x}_1, {\bf x}_2\right\}$ be a basis for a subspace $W$ of $\mathbb{R}^3$. Construct an orthogonal basis for $W$.

**Solution.**

We start with ${\bf x}_1$ and construct a vector orthogonal to it by taking the component of ${\bf x}_2$ that is orthogonal to ${\bf x}_1$ (see the figure below)


<center> <img src="data:image/png;base64,  
 " width="40%"> </center>

Let ${\bf v}_1={\bf x}_1$ and ${\bf v}_2={\bf x}_2-proj_{{\bf v}_1}({\bf x}_2)$.
Then,
$${\bf v}_2={\bf x}_2-proj_{{\bf v}_1}({\bf x}_2)= {\bf x}_2 - \frac{ {\bf x}_2\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1=\begin{bmatrix} -1 \\ 1 \\ 2 \end{bmatrix} - \frac{7}{14} \begin{bmatrix} 1 \\ 2\\3\end{bmatrix}=\begin{bmatrix} -3/2 \\ 0 \\ 1/2 \end{bmatrix} $$

The set $\{ {\bf v}_1, {\bf v}_2 \}$ is orthogonal and hence linearly independent. Therefore, it is an orthogonal basis for $W$. 

Since scaling a basis vector by a nonzero scalar does not affect orthogonality, we may scale the vectors to simplify their entries. For instance, multiplying ${\bf v}_2$ by $2$ eliminates the fractions and gives the cleaner orthogonal basis $$ \left\{ \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix}, \begin{bmatrix} -3 \\ 0 \\ 1\end{bmatrix} \right\}$$

---

In the next example, we apply the same technique to a basis consisting of three vectors.

---
**<font color="red">Example 2</font>**

Let ${\bf x}_1 =\begin{bmatrix} 1 \\ 2 \\ 0 \\ 1 \end{bmatrix}$, ${\bf x}_2=\begin{bmatrix} -1 \\ 1 \\ 0 \\ 1 \end{bmatrix}$ and ${\bf x}_3=\begin{bmatrix} 0 \\ 1 \\ 2 \\3 \end{bmatrix}$. Let $\mathcal{B}=\left\{{\bf x}_1, {\bf x}_2, {\bf x}_3\right\}$ be a basis for a subspace $W$ of $\mathbb{R}^4$. Construct an orthogonal basis for $W$.

**Solution.**

We perform the orthogonalization one vector at a time.

**Step 1.** 
Let ${\bf v}_1={\bf x}_1$ and let $W_1=span({\bf v}_1)=span({\bf x}_1)$.

**Step 2.** 
As in Example 1, we construct ${\bf v}_2$ by taking the component of ${\bf x}_2$ orthogonal to $W_1$.

$${\bf v}_2={\bf x}_2-proj_{{\bf v}_1}({\bf x}_2)= {\bf x}_2 - \frac{ {\bf x}_2\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1=\begin{bmatrix} -1 \\ 1 \\ 0 \\ 1 \end{bmatrix} - \frac{2}{6} \begin{bmatrix} 1 \\ 2 \\ 0 \\ 1 \end{bmatrix}=\begin{bmatrix} -4/3 \\ 1/3 \\ 0\\ 2/3 \end{bmatrix} \rightarrow \begin{bmatrix} -4 \\ 1 \\ 0\\ 2 \end{bmatrix} $$

We scaled the resulting vector to obtain simpler entries, which will make the next step easier.

Let $W_2=span({\bf v}_1, {\bf v}_2)=span({\bf x}_1, {\bf x}_2)$.

**Step 3.** 
Now we use ${\bf x}_3$ to construct a vector orthogonal to $W_2$. More precisely, we take the component of ${\bf x}_3$ orthogonal to $W_2$.

$${\bf v}_3={\bf x}_3-proj_{W_2}({\bf x}_3)= {\bf x}_3-proj_{{\bf v}_1}({\bf x}_3)-proj_{{\bf v}_2}({\bf x}_3)={\bf x}_3 - \frac{ {\bf x}_3\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1 - \frac{ {\bf x}_3\cdot {\bf v}_2}{{\bf v}_2 \cdot {\bf v}_2}{\bf v}_2$$
$$=\begin{bmatrix} 0 \\ 1 \\ 2 \\3 \end{bmatrix}-\frac{5}{6}\begin{bmatrix} 1 \\ 2 \\ 0 \\ 1 \end{bmatrix} - \frac{7}{21}\begin{bmatrix} -4 \\ 1 \\ 0\\ 2 \end{bmatrix}=\begin{bmatrix} 1/2 \\ -1 \\ 2 \\ 3/2 \end{bmatrix} \rightarrow \begin{bmatrix} 1 \\ -2 \\ 4\\ 3 \end{bmatrix}$$

The set $\left\{{\bf v}_1, {\bf v}_2, {\bf v}_3\right\}$ is orthogonal and therefore linearly independent. Moreover, $span({\bf v}_1, {\bf v}_2, {\bf v}_3)=span({\bf x}_1, {\bf x}_2, {\bf x}_3)=W$. Hence, $\left\{{\bf v}_1, {\bf v}_2, {\bf v}_3\right\}$ is an orthogonal basis for $W$.

The formula for projecting onto a subspace by summing projections onto individual basis vectors is valid only when the basis is orthogonal.

<strong>🏛️ Theorem </strong><br>
    Let $\{ {\bf x}_1, \cdots, {\bf x}_k \}$ be a basis for a subspace $W$ of $\mathbb{R}^n$. Define <br>
${\bf v}_1 = {\bf x}_1$ <br>

${\bf v}_2 = {\bf x}_2 - \frac{ {\bf x}_2\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1$<br>

${\bf v}_3= {\bf x}_3 - \frac{ {\bf x}_3\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1 - \frac{ {\bf x}_3\cdot {\bf v}_2}{{\bf v}_2 \cdot {\bf v}_2}{\bf v}_2$<br>

$\vdots$ <br>

${\bf v}_k= {\bf x}_k - \frac{ {\bf x}_k\cdot {\bf v}_1}{{\bf v}_1 \cdot {\bf v}_1}{\bf v}_1 - \frac{ {\bf x}_k\cdot {\bf v}_2}{{\bf v}_2 \cdot {\bf v}_2}{\bf v}_2 - \cdots - \frac{ {\bf x}_k\cdot {\bf v}_{k-1}}{{\bf v}_{k-1} \cdot {\bf v}_{k-1}}{\bf v}_{k-1}$<br>

Then $\{ {\bf v}_1, \cdots, {\bf v}_k \}$ is an orthogonal basis for $W$. Moreover, $span( {\bf v}_1, \cdots, {\bf v}_j )=span( {\bf x}_1, \cdots, {\bf x}_j )$ for every $1 \leq j \leq k$.
</div>



Once we have an orthogonal basis, it is easy to construct an orthonormal basis. We simply normalize each basis vector.

---
**<font color="red">Example 3 (continuation of Example 2)</font>**

Using the vectors from Example 2, construct an orthonormal basis.

**Solution.**

It is convenient to use the scaled vectors from Example 2 when computing their norms. The orthogonal basis obtained there is 
$$ {\bf v}_1=\begin{bmatrix} 1 \\ 2 \\ 0 \\ 1 \end{bmatrix},{\bf v}_2= \begin{bmatrix} -4 \\ 1 \\ 0\\ 2 \end{bmatrix}, {\bf v}_3=\begin{bmatrix} 1 \\ -2 \\ 4\\ 3 \end{bmatrix}$$

Normalizing each vector gives the following orthonormal basis:

$$ {\bf u}_1=\frac{{\bf v}_1}{||{\bf v}_1||}=\frac{1}{\sqrt{6}}\begin{bmatrix} 1 \\ 2 \\ 0 \\ 1 \end{bmatrix}=\begin{bmatrix} 1/\sqrt{6} \\ 2/\sqrt{6} \\ 0 \\ 1/\sqrt{6} \end{bmatrix}$$
$$ {\bf u}_2=\frac{{\bf v}_2}{||{\bf v}_2||}=\frac{1}{\sqrt{21}}\begin{bmatrix} -4 \\ 1 \\ 0\\ 2 \end{bmatrix}=\begin{bmatrix} -4/\sqrt{21} \\ 1/\sqrt{21} \\ 0\\ 2/\sqrt{21} \end{bmatrix}$$
$$ {\bf u}_3=\frac{{\bf v}_3}{||{\bf v}_3||}=\frac{1}{\sqrt{30}}\begin{bmatrix} 1 \\ -2 \\ 4\\ 3 \end{bmatrix}=\begin{bmatrix} 1/\sqrt{30} \\ -2/\sqrt{30}  \\ 4/\sqrt{30} \\ 3/\sqrt{30}  \end{bmatrix}$$

<b>QR Factorization</b>
</div></font> 

If we apply the Gram-Schmidt process to the columns of an $m\times n$ matrix $A$ with linearly independent columns, we obtain a very useful factorization of $A$ into the product of a matrix $Q$ with orthonormal columns and an upper triangular matrix $R$. This is called **QR factorization**, and it has many applications in computational linear algebra, some of which we will discuss in this and subsequent chapters.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem </strong><br> If $A$ is an $m \times n$ matrix with linearly independent columns, then $A$ can be factored as $A=QR$, where $Q$ is an $m \times n$ matrix whose columns form an orthonormal basis for $Col(A)$ and $R$ is an invertible $n \times n$ upper triangular matrix with positive diagonal entries.
</div>


**Proof.**

Let $A=\begin{bmatrix} {\bf a}_1 \cdots {\bf a}_n \end{bmatrix}$. Since the columns of $A$ are linearly independent, $\{ {\bf a}_1, \cdots, {\bf a}_n \}$ is a basis for $Col(A)$. Apply the Gram-Schmidt process followed by normalization to $\{ {\bf a}_1, \cdots, {\bf a}_n \}$ to obtain an orthonormal basis $\{ {\bf q}_1, \cdots, {\bf q}_n \}$  for $Col(A)$. Let $Q=\begin{bmatrix} {\bf q}_1 \cdots {\bf q}_n \end{bmatrix}$. The columns of $Q$ clearly form an orthonormal set.

Furthermore, for each $i=1, \cdots, n$, $$span({\bf a}_1, \cdots, {\bf a}_i)=span({\bf q}_1, \cdots, {\bf q}_i)$$ 

Therefore, there exist scalars $r_{1i}, \cdots, r_{ii}$ such that $${\bf a}_i=r_{1i}{\bf q}_1 + \cdots + r_{ii}{\bf q}_i$$

This means

<div style="display: flex; justify-content: center;">
    <div style="text-align: left;">
        ${\bf a}_1 = r_{11}{\bf q}_1 $<br>
        ${\bf a}_2= r_{12}{\bf q}_1+r_{22}{\bf q}_2 $<br>
        ${\bf a}_3= r_{13}{\bf q}_1+r_{23}{\bf q}_2+r_{33}{\bf q}_3$<br>
        $\text{  }\vdots$<br>
        ${\bf a}_n= r_{1n}{\bf q}_1+r_{2n}{\bf q}_2+ \cdots + r_{nn}{\bf q}_n$
    </div>
</div>

We may choose the sign of each ${\bf q}_i$, along with the corresponding coefficients, so that $r_{ii} \geq 0$ for every $i$.

Therefore,

$$A=\begin{bmatrix} {\bf a}_1 \cdots {\bf a}_n \end{bmatrix} = \begin{bmatrix} {\bf q}_1 \cdots {\bf q}_n \end{bmatrix} \begin{bmatrix} r_{11} & r_{12} & \cdots & r_{1n} \\ 0 & r_{22} & \cdots & r_{2n} \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \cdots & r_{nn} \end{bmatrix}=QR$$



Finally, suppose that $r_{ii} =  0$ for some $i$. Then ${\bf a}_i$ belongs to $span({\bf q}_1, \cdots, {\bf q}_{i-1})=span({\bf a}_1, \cdots, {\bf a}_{i-1})$. This is impossible because the columns of $A$ are linearly independent. Therefore, $r_{ii}>0$ for every $i$, and so $R$ is an upper triangular matrix with positive diagonal entries.

---
**<font color="red">Example 4</font>**

Find the QR factorization of $A=\begin{bmatrix} 1 & -1 & 0 \\ 2 & 1 & 1 \\ 0 & 0 & 2 \\ 1 & 1 & 3 \end{bmatrix}$

**Solution.**

The columns of $A$ are the vectors used in Examples 2 and 3. Applying the Gram-Schmidt process to these columns and then normalizing the resulting vectors gives the orthonormal basis $$Q=\begin{bmatrix} {\bf u}_1 & {\bf u}_2 & {\bf u}_3 \end{bmatrix} =\begin{bmatrix} 1/\sqrt{6} & -4/\sqrt{21}& 1/\sqrt{30} \\ 2/\sqrt{6} &1/\sqrt{21}&-2/\sqrt{30}\\ 0 &0& 4/\sqrt{30}\\ 1/\sqrt{6}&2/\sqrt{21}&3/\sqrt{30} \end{bmatrix}$$

Next, we determine $R$ so that $A=QR$. Since $Q$ has orthonormal columns, $Q^TQ=I$. Multiplying both sides of $A=QR$ by $Q^T$ gives
$$Q^TA=Q^TQR=IR=R$$

Therefore,

$$R=Q^TA=\begin{bmatrix} 1/\sqrt{6} & 2/\sqrt{6} & 0 & 1/\sqrt{6} \\-4/\sqrt{21} & 1/\sqrt{21} & 0& 2/\sqrt{21}\\  1/\sqrt{30} & -2/\sqrt{30} & 4/\sqrt{30}& 3/\sqrt{30}  \end{bmatrix} \begin{bmatrix} 1 & -1 & 0 \\ 2 & 1 & 1 \\ 0 & 0 & 2 \\ 1 & 1 & 3 \end{bmatrix}=\begin{bmatrix} \sqrt{6} & 2/\sqrt{6} & 5/\sqrt{6} \\ 0& 7/\sqrt{21} & 7/\sqrt{21} \\ 0 & 0 & 15/\sqrt{30}  \end{bmatrix}$$

---

Although the QR factorization can be derived using the Gram-Schmidt process, modern numerical algorithms rarely compute it this way. The classical Gram-Schmidt process is susceptible to rounding errors, especially when the columns of $A$ are nearly linearly dependent.

Instead, most numerical software computes the QR factorization using **Householder reflections**. Although this method can require more computations, it is significantly more numerically stable and produces more accurate results, particularly for nearly linearly dependent columns.

<b>QR Factorization in Python</b>
</div></font>  

In Python, we can compute the QR factorization of a matrix using `la.qr()`. This function returns two matrices, $Q$ and $R$, such that $A=QR$ and $Q^TQ=I$. By default, `la.qr()` returns a square matrix $Q$. However, when $A$ has more rows than columns, some columns of $Q$ may be unnecessary for reconstructing $A$. To avoid these extra columns, we use `mode='economic'`, which returns a smaller matrix $Q$. This is called the reduced QR factorization. 

When the columns of $A$ are linearly independent, the columns of $Q$ form an orthonormal basis for $Col(A)$.

---
**<font color="red">Example 5</font>**

Use Python to find an orthonormal basis for $Col(A)$, where $A=\begin{bmatrix} 1 & -1 & 0 \\ 2 & 1 & 1 \\ 0 & 0 & 2 \\ 1 & 1 & 3 \end{bmatrix}$.

**Solution.**

```python
A=np.array([[1, -1, 0],[2,1,1],[0,0,2],[1,1,3]])
Q, R=la.qr(A,mode='economic')
print(Q, '\n')
print(R)
```

The three columns of $Q$ form an orthonormal basis for $Col(A)$, as expected from Example 4.

The signs of the columns of $Q$ and the corresponding rows of $R$ may differ from those obtained in Example 4. SciPy does not require the diagonal entries of $R$ to be positive, but the factorization $A=QR$ is still valid.
</div> 

---

The Householder reflections method used by `la.qr()` also works when the columns of $A$ are linearly dependent. In this case, whether the columns of $Q$ form a basis for $Col(A)$ depends on the dimension of the column space.

We will consider two examples.

---
**<font color="red">Example 6</font>**

Find orthonormal bases for $Col(A)$ and $Row(A)$, where $A=\begin{bmatrix}1 & 1 & 1 & 1 \\ 2 & 1 & 0 & 0 \end{bmatrix}$

**Solution.**


```python
A=np.array([[1,1,1,1],[2,1,0,0]])
Q, R=la.qr(A,mode='economic')
print(Q)
print(f'\n Rank of A is {np.linalg.matrix_rank(A)}')
```

Although the columns of $A$ are linearly dependent, $A$ has a pivot in every row. Thus, $Col(A)=\mathbb{R}^2$, and the two columns of $Q$ form an orthonormal basis for $Col(A)$.

Since $Row(A)=Col(A^T)$, we can obtain an orthonormal basis for $Row(A)$ by applying QR factorization to $A^T$. 

```python
Q, R=la.qr(A.T,mode='economic')
print(Q)
```

The columns of $A^T$ are linearly independent, so the columns of $Q$ form an orthonormal basis for $Row(A)$.

---

**<font color="red">Example 7</font>**

Find an orthonormal basis for $Col(A)$, where $A=\begin{bmatrix}1 & 1 & 1 & 1 \\ 2 & 1 & 0 & 0 \\ 1 & 1 & 1 & 1 \end{bmatrix}$

**Solution.**


```python
A=np.array([[1,1,1,1],[2,1,0,0],[1,1,1,1]])
Q, R=la.qr(A.T,mode='economic')
print(Q)
print(f'\n Rank of A is {np.linalg.matrix_rank(A)}')
```

In this case, $A$ has only two pivots, so $dim(Col(A))=2$. However, $Q$ has three orthonormal columns. Therefore, not all of its columns can belong to $Col(A)$. To identify an orthonormal basis, we can use QR factorization with column pivoting by specifying `pivoting=True`, which rearranges the columns of $A$ during the factorization. 


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #4c72b0;
    background-color: rgba(10,150,250,0.16);
    margin: 0.4em 0;
">
    
<strong>📌 Note:</strong> When `pivoting=True`, `la.qr()` returns three objects: $Q$, $R$, and $P$. Here, $P$ is an array specifying how the columns of $A$ are rearranged, so that `A[:, P] = Q @ R`. Column pivoting helps identify the linearly independent directions in the column space.
</div> 

```python
Q, R, P = la.qr(A, mode='economic', pivoting=True)

print(Q)
print('\n', P)
```

With column pivoting, the first two columns of $Q$ form an orthonormal basis for $Col(A)$, since $A$ has two pivots.

```python
orthonormal_basis = Q[:, :2]
print(f' Orthonormal Basis:\n {orthonormal_basis}')
```
