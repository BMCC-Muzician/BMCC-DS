2.1 Introduction to Transformations
</div></font>

So far, we have viewed matrix–vector multiplication as a way to form linear combinations of vectors, or as an efficient way to compute dot products with multiple vectors at once. However, matrices can also be used to transform vectors, acting as a type of function $${\bf y}=T({\bf x})=A{\bf x}$$ where independent variable ${\bf x}$ and dependent variable ${\bf y}$ are both vectors. 

To formalize this idea, we introduce the concept of a **transformation**. This definition is very similar to the definition of a function. The main difference is that we are now allowed to use vectors as inputs and outputs.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong><br> A <strong>transformation</strong> (or <strong>mapping</strong>, or <strong>map</strong>) $T$ from $\mathbb{R}^n$ to $\mathbb{R}^m$ is a rule that assigns to each vector ${\bf x}$ in $\mathbb{R}^n$ a unique vector $T({\bf x})$ in $\mathbb{R}^m$. The <strong>domain</strong> of $T$ is $\mathbb{R}^n$, and the <strong>codomain</strong> is $\mathbb{R}^m$. We indicate this as $$T:\mathbb{R}^n \rightarrow \mathbb{R}^m$$ Given vector ${\bf x}$ in the domain of $T$, the vector $T({\bf x})$ is called an <strong>image</strong> of ${\bf x}$ under $T$. The set of all images $T({\bf x})$, as ${\bf x}$ varies throughout the domain, is called the <strong>range</strong> of $T$. Given a vector ${\bf y}$ in range of $T$, a vector ${\bf x}$ such that $T({\bf x})={\bf y}$ is called a <strong>pre-image</strong> of ${\bf y}$.
</div>



---
**<font color="red">Example 1</font>**

Let $A=\begin{bmatrix} 1 & 2 & 3 \\  0 & -1 & 2 \end{bmatrix}$ and $T({\bf x})=A{\bf x}$. Since ${\bf x}$ must have as many entries as there are columns of $A$, the domain of $T$ is $\mathbb{R}^3$. After multiplying $A$ by ${\bf x}$, we get a vector that has as many entries as there are rows of $A$, and so codomain is $\mathbb{R}^2$. 

Let ${\bf v}=\begin{bmatrix} 1 \\4 \\ -2 \end{bmatrix}$. Then 

$$T({\bf v})=A{\bf v}=\begin{bmatrix} 1 & 2 & 3 \\  0 & -1 & 2 \end{bmatrix}\begin{bmatrix} 1 \\4 \\ -2 \end{bmatrix}=\begin{bmatrix} 3 \\-8  \end{bmatrix}$$

The resulting vector $\begin{bmatrix} 3 \\-8  \end{bmatrix}$ is the image of ${\bf v}$ under $T$.

---

Transformations that can be written in the form $T({\bf x})=A{\bf x}$ for some matrix $A$ are called **matrix transformations**. These will be the most important type of transformations for us. Note that not every transformation is a matrix transformation. 

---
**<font color="red">Example 2</font>**

Let $T:\mathbb{R}^3 \rightarrow \mathbb{R}$ be the transformation defined by $T({\bf x})=||{\bf x}||$. 

1. What is the range of $T$? Describe all pre-images of $1$.
2. Why can't $T$ be written as a matrix transformation?
   
**Solution.**

1. Since norm is always non-negative, the range of $T$ is $[0, \infty)$. 

The pre-images of $1$ are all vectors whose norm is equal to $1$. Geometrically, this is the set of all points at distance $1$ from the origin in $\mathbb{R}^3$, which forms a sphere of radius $1$.


2. Consider the image of a $-{\bf x}$ under $T$:

$$T(-{\bf x})=||-{\bf x}||=||{\bf x}||$$

However, if $T$ were a matrix transformation  $T({\bf x})=A{\bf x}$, then
$$T(-{\bf x})=A(-{\bf x})=-A{\bf x}=-T({\bf x})=-||{\bf x}||$$

This would imply that $||{\bf x}||=-||{\bf x}||$, which is true only if ${\bf x}={\bf 0}$. Since this is not true for all vectors ${\bf x}$, $T$ is not a matrix transformation.

<b>Linear Transformations</b>
</div></font> 

The two fundamental properties of matrix-vector multiplication are preservation of addition and preservation of scalar multiplication:
$$A({\bf v}+{\bf u})=A{\bf v}+A{\bf u}$$
$$A(c{\bf v})=c(A{\bf v})$$

These properties motivate one of the most important concepts in linear algebra: **linear transformations**.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong> <br> A transformation $T$ is called <strong>linear</strong> if for any vectors ${\bf u}$, ${\bf v}$ in the domain of $T$, and any scalar $c$, the following holds:<br>
<strong>1.</strong> $T({\bf u}+{\bf v})=T({\bf u})+T({\bf v})$<br>
<strong>2.</strong> $T(c{\bf u})=cT({\bf u})$
</div> 


It is easy to see that every matrix transformation must be a linear transformation. Therefore, the transformation in the Example 1 is linear. However, the transformation $T({\bf x})=||{\bf x}||$ in Example 2 is not linear. For instance, it does not preserve scalar multiplication. For example, let $c=-2$, then

$$T(-2{\bf v})=||-2{\bf v}||=2||{\bf v}||$$

but

$$-2T({\bf v})=-2||{\bf v}||$$

Since at least one property of linearity fails, $T$ is not a linear transformation.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong> <br> If $T$ is a linear transformation, then<br>
<strong>1.</strong> $T({\bf 0})={\bf 0}$<br>
<strong>2.</strong> $T(c{\bf u}+d{\bf v})=cT({\bf u})+dT({\bf v})$, for any vectors ${\bf u}$ and ${\bf v}$ in domain of $T$, and any scalars $c$ and $d$.
</div>


The first property can be used as quick way to show that certain transformations are not linear.


<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #dd8452;
    background-color: rgba(250,100,60,0.16);
    margin: 0.4em 0;
">
<strong>⚠️Warning</strong><br> If the transformation $T$ does not satisfy $T({\bf 0})={\bf 0}$, then $T$ is not linear. However, if $T({\bf 0})={\bf 0}$, then it is not enough to determine whether $T$ is linear or not. For instance, the transformation in Example 2 is not linear, even though $T({\bf 0})=||{\bf 0}||=0$.
</div>





In the next example, we will use an alternative notation for transformations. We can always rewrite it using vector notation if it is more convenient. So, $$T(x,y)=(x+1, x+y)$$ is the same as  $$T\left(\begin{bmatrix} x\\y\end{bmatrix}\right)=\begin{bmatrix}x+1\\ x+y\end{bmatrix}$$

**<font color="red">Example 3</font>**

Show that $T(x,y)=(x+1, x+y)$ is not a linear transformation.

**Solution.**

Evaluate at the zero vector:
$$T(0,0)=(1,0)$$
Since the output is not the zero vector, $T$ is not linear.

---

**<font color="red">Example 4</font>**

Show $T(x,y)=(x+y, y^2)$ is not linear.

**Solution.**

First, note that $T(0,0)=(0,0)$, so a zero-vector test does not detect non-linearity. We must check the definition.

Let's check whether $T$ preserves addition. On one hand:

$$T(x_1+x_2, y_1+y_2)=(x_1+x_2+ y_1+y_2, (y_1+y_2)^2)=(x_1+ y_1+x_2+y_2, y_1^2+2y_1y_2+y_2^2)$$

On the other hand:

$$T(x_1, y_1)+T(x_2, y_2)=(x_1+y_1, y_1^2)+(x_2+y_2, y_2^2)=(x_1+ y_1+x_2+y_2, y_1^2+y_2^2)$$

And so $T(x_1+x_2, y_1+y_2) \neq T(x_1, y_1)+T(x_2, y_2)$ due to $2y_1y_2$ term. Therefore, $T$ is not a linear transformation.

Once a transformation fails one of the linearity properties, there is no need to check the other. However, for completeness, we can also check the second property. It works similarly. 

$$T(cx,cy)=(cx+cy, (cy)^2)=(cx+cy, c^2y^2)$$
$$cT(x,y)=c(x+y, y^2)=(cx+cy, cy^2)$$

Thus, the second property also fails.

<b>Standard Matrix for Transformation</b>
</div></font> 


It is natural to ask three important questions:

* How can we tell whether a transformation is a matrix transformation?
* If a transformation is a matrix transformation, how do we find its associated matrix?
* Are there linear transformations that are not matrix transformations?

The following theorem answers these questions.



<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️Theorem</strong> <br>
<strong>1.</strong> Every matrix transformation is a linear transformation.<br>
<strong>2.</strong> Every linear transformation $T$ from $\mathbb{R}^n$ to $\mathbb{R}^m$ is a matrix transformation. Specifically, $T({\bf x})=M{\bf x}$, where $$M=\begin{bmatrix} T({\bf e}_1) & \cdots &T({\bf e}_n) \end{bmatrix}$$
</div>


**Proof.**

1. The first statement is obvious, since it follows directly from the fact that matrix–vector multiplication satisfies both properties of linearity.

2. Let $\bf x$ be a vector in $\mathbb{R}^n$. We can use the standard basis $\{ {\bf e}_1, \cdots, {\bf e}_n \}$ to write

$${\bf x}=x_1{\bf e}_1+x_2{\bf e}_2 + \cdots+ x_n{\bf e}_n$$

Since $T$ is linear,

$$T({\bf x})=T(x_1{\bf e}_1+x_2{\bf e}_2 + \cdots+ x_n{\bf e}_n)=x_1T({\bf e}_1)+\cdots+x_nT({\bf e}_n)=\begin{bmatrix} T({\bf e}_1) & \cdots &T({\bf e}_n) \end{bmatrix} \begin{bmatrix} x_1 \\ \vdots \\ x_n \end{bmatrix} = M{\bf x}$$

---

The matrix $M$ in the above theorem is called the **standard matrix of the linear transformation $T$**. This is a fundamental idea: many transformations can be understood by studying their standard matrices. The next two examples explore this.

---
**<font color="red">Example 5</font>**

Let $\mathcal{B}=\{ {\bf b}_1, \cdots, {\bf b}_n \}$ be a basis for $\mathbb{R}^n$ and $T({\bf x})=[{\bf x}]_{\mathcal{B}}$. Show that $T$ is linear  and find its standard matrix.

**Solution**

The fact that $T$ is a linear transformation follows from the properties of coordinate vectors. For any vectors $\bf u$, $\bf v$ in $\mathbb{R}^n$ and any scalar $c$:

1. $T({\bf u}+{\bf v})=[{\bf u}+{\bf v}]_{\mathcal{B}}=[{\bf u}]_{\mathcal{B}}+[{\bf v}]_{\mathcal{B}}=T({\bf u})+T({\bf v})$
2. $T(c{\bf u})=[c{\bf u}]_{\mathcal{B}}=c[{\bf u}]_{\mathcal{B}}=cT({\bf u})$


To answer the second question, notice that $T$ converts coordinates from the standard basis $\mathcal{E}$ to the basis $\mathcal{B}$, which is exactly what change-of-coordinate matrix $\underset{\mathcal{B}\leftarrow\mathcal{E}}{P}$ does. Therefore, $$T({\bf x})=\underset{\mathcal{B}\leftarrow\mathcal{E}}{P} {\bf x}$$


---
**<font color="red">Example 6</font>**

Let $T: \mathbb{R}^2 \rightarrow \mathbb{R}^2$ be the transformation that rotates every vector by an angle $\theta$ (in radians) counterclockwise  with respect to the origin. Show that $T$ is linear and find its standard matrix.

**Solution**

It is easy to see that $T({\bf u}+{\bf v})=T({\bf u})+T({\bf v})$ by considering rotation of parallelogram created by the origin, ${\bf u}$, ${\bf v}$ and ${\bf u}+{\bf v}$.

Similarly, to see that $T(c{\bf u})=cT({\bf u})$ we can observe that scaling and then rotating a vector is equivalent to rotating and then scaling it.

 " width="70%"> </center>

To find the standard matrix for this transformation, we need to find images of the standard basis vectors under $T$. This requires a bit of trigonometry.

<center> <img src="data:image/png;base64,  
 " width="35%"> </center>

$$T({\bf e}_1)=T\left(\begin{bmatrix} 1 \\0 \end{bmatrix}\right)=\begin{bmatrix} \cos(\theta) \\ \sin(\theta)\end{bmatrix}$$
$$T({\bf e}_2)=T\left(\begin{bmatrix} 0 \\1 \end{bmatrix}\right)=\begin{bmatrix} -\sin(\theta) \\ \cos(\theta)\end{bmatrix}$$

And so, $$T({\bf x})=\begin{bmatrix} \cos(\theta) & -\sin(\theta) \\ \sin(\theta) & \cos(\theta)\end{bmatrix}{\bf x}$$

This is a powerful technique, as it allows us to move between different interpretations of a matrix: the coefficient matrix of a linear system or the standard matrix of a linear transformation.
</div>

<b>Other Properties of Transformations</b>
</div></font> 

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong> <br>Let $T:\mathbb{R}^n \rightarrow \mathbb{R}^m$ be a transformation. The <strong>kernel</strong> of $T$ is the set of vectors ${\bf x} \in \mathbb{R}^n$ such that  $T({\bf x})={\bf 0}$. It is denoted by $ker(T)$.
</div>


If $T({\bf x})=A{\bf x}$ is a matrix transformation then
$$range(T)=Col(A) \text{ \quad and \quad }ker(T)=Nul(A)$$


This is an important observation. Since every linear transformation can be written as a matrix transformation, it follows that for any linear transformation, both the kernel and the range are vector subspaces.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong> <br>A transformation $T:\mathbb{R}^n \rightarrow \mathbb{R}^m$ is <strong>onto</strong> (or <strong>surjective</strong>) if every vector ${\bf b} \in \mathbb{R}^m$ has at least one pre-image in $\mathbb{R}^n$.
Equivalently, $range(T)=\mathbb{R}^m$.
</div>




If $T({\bf x})=A{\bf x}$ is a onto matrix transformation, then $Col(A)=\mathbb{R}^m$.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Definition</strong> <br>A transformation $T:\mathbb{R}^n \rightarrow \mathbb{R}^m$ is <strong>one-to-one</strong> (or <strong>injective</strong>) if every vector ${\bf b} \in \mathbb{R}^m$ has at most one pre-image in $\mathbb{R}^n$.

Equivalently, $T$ is one-to-one if, for any two distinct vectors $\bf x$ and $\bf y$ in its domain, $T({\bf x})\neq T({\bf y})$
</div>


Suppose $T$ is a one-to-one linear transformation. Since $T({\bf 0})={\bf 0}$, no other vector can be mapped to the zero vector. Therefore, $ker(T)=\{ {\bf 0}\}$. The reverse of this statement is also true. If we know that $ker(T)=\{ {\bf 0}\}$ and $T$ is linear, then $T$ must be one-to-one:

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 1.6;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong> <br>Let $T:\mathbb{R}^n \rightarrow \mathbb{R}^m$ be a linear transformation. Then $T$ is one-to-one if and only if $ker(T)=\{{\bf 0}\}$
 (this is equivalent to saying that $T({\bf x})={\bf 0}$ has only trivial solution)
</div>

**Proof.**

Here, we will prove only one direction:  if $ker(T)=\{ {\bf 0}\}$ and $T$ is linear, then $T$ must be one-to-one.

Suppose $T$ is not one-to-one. Then there are two distinct vectors ${\bf u}$ and $\bf v$ such that $T({\bf u})=T({\bf v})$

Since ${\bf u}$ and $\bf v$ are distinct, ${\bf u}-{\bf v} \neq {\bf 0}$. And since only zero vector is the kernel, $$T({\bf u}-{\bf v})\neq {\bf 0}$$

However, on the other hand, by linearity $$T({\bf u}-{\bf v})=T({\bf u})-T({\bf v})={\bf 0}$$

This is a contradiction. Therefore, $T$ must be one-to-one.

---

The connection between injectivity/surjectivity and vector subspaces is very useful, because we can study these properties using the standard matrix.

<div style="
    font-family: Verdana, sans-serif;
    font-size: 14px;
    line-height: 2;
    margin: 0.8em 0;
    padding: 0.75em 1em;
    border-radius: 6px;
    border-left: 4px solid #2c3e50;
    border-left-color: #55a868;
    background-color: rgba(10,150,50,0.16);
    margin: 0.4em 0;
">
<strong>🏛️ Theorem</strong> <br>
Let $T:\mathbb{R}^n \rightarrow \mathbb{R}^m$ be a linear transformation, and let $A$ be the standard matrix for $T$. Then<br>
<strong>1.</strong> $T$ is onto if and only if the columns of $A$ span $\mathbb{R}^m$, which is equivalent to $A$ having a pivot in every row.<br>
<strong>2.</strong> $T$ is one-to-one if and only if the columns of $A$ are linearly independent, which is equivalent to $A$ having a pivot in every column.
</div>

**<font color="red">Example 7</font>**

Let $T(x_1, x_2)=(2x_1-3x_2, x_1-x_2, 2x_1+5x_2)$. 

1. Determine whether $T$ is one-to-one and/or onto.
2. Find $range(T)$ and $ker(T)$ and state their dimensions.

**Solution**

1. We first express $T$ as a matrix transformation.

$$T\left(\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}\right)=\begin{bmatrix} 2x_1-3x_2 \\ x_1-x_2 \\ 2x_1+5x_2\end{bmatrix}=\begin{bmatrix} 2 & -3 \\ 1 & -1 \\ 2 & 5 \end{bmatrix}\begin{bmatrix}  x_1 \\ x_2 \end{bmatrix}$$

Now we can row reduce the matrix to find the pivot positions.

$$\begin{bmatrix} 2 & -3 \\ 1 & -1 \\ 2 & 5 \end{bmatrix} \sim \cdots \sim \begin{bmatrix} 1 & -1 \\ 0 & 1 \\ 0 & 0 \end{bmatrix}$$

Since we have a pivot in every column, $T$ is one-to-one. However, since we don't have a pivot in every row, $T$ is not onto.

2. The kernel of a matrix transformation is simply the null space of its standard matrix. Since $T$ is one-to-one, $ker(T)=\{ {\bf 0}\}$. And so, $$dim(ker(T))=0$$
   The range of a matrix transformation is the column space of the standard matrix. In our case, the columns of the standard matrix are linearly independent, and so they form a basis for the range:
   $$range(T)=span\left( \begin{bmatrix} 2  \\ 1 \\ 2 \end{bmatrix}, \begin{bmatrix}  -3 \\  -1 \\ 5 \end{bmatrix} \right)$$
   And so, $$dim(range(T))=2$$

<b>Visualizing Transformations</b>
</div></font> 

To understand linear transformations geometrically, it is helpful to introduce the idea of a **flat**.

A **flat** is a general term for geometric objects such as points, lines, planes, and their higher-dimensional analogues. The dimension of a flat is defined in the natural way:
* A point has dimension $0$
* A line has dimension $1$
* A plane has dimension $2$ 
* And so on

Here are some of the geometric properties of linear transformation:

1. **Origin is preserved**

    The image of the origin must be the origin.

2. **Dimension does not increase**
   
    An image of a flat of dimension $n$ is a flat of dimension at most $n$. For example, the image of a line must be a line or a point. A plane can remain a plane, become a line, or collapse to a point.

   
3. **Parallel structure is preserved**

    Images of parallel flats are either parallel or coincide (collapse onto each other). For example, a square grid in the plane may become a skewed grid, or it may collapse into a line.
    
4. **Pre-images form parallel flats**

    The pre-images of points in the range have a very specific structure. The pre-image of ${\bf 0}$ is the kernel (which is a subspace), and the pre-image of any other vector in the range is a shifted copy of the kernel. In particular, all pre-images have the same dimension and are parallel to each other.


Intuitively, we can think of linear transformations as 
* stretching
* compressing
* rotating
* shearing
* flattening
* any combination of the above

but always in a way that preserves the above properties.
